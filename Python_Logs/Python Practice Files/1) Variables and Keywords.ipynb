# Variables and Keywords:

A Python variable is a reserved memory location to store values. In other words, a variable in a python program gives data to the computer for processing.

## Variables:

```python
a = 'hello world'
```

```python
print(a) # To Print Variable Value.
```

```python
print(type(a)) # To check Datatype of Variable.
```

```python
my_age = 33
print(my_age)
print(type(my_age))
```

```python
# Different Types of Data:
print(type(23))
print(type(23.67))
print(type('Hello'))
print(type(True))
```

```python
score = 89.5
print(score)
print(type(score))
```

```python
var = 'Hello World!!'
print(var)
print(type(var))
```

```python
a = 10.5
b = '10.5'
print(a, b)
print(type(a), type(b))
```

```python
x = 'True'
y = True
print(x, y)
print(type(x), type(y))
```

```python
# Variable Names can not start with Digits:
5a = 10
```

```python
# Variable Names are Case Sensitive:
a = 10
A = 15
print(a, A)
```
