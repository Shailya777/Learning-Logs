### Concat and Merge Dataframe:

```python
df1 = pd.DataFrame({'A' : ['A0', 'A1', 'A2', 'A3'],
                   'B' : ['B0', 'B1', 'B2', 'B3'],
                   'C' : ['C0', 'C1', 'C2', 'C3'],
                   'D' : ['D0', 'D1', 'D2', 'D3']})

df2 = pd.DataFrame({'A' : ['A4', 'A5', 'A6', 'A7'],
                   'B' : ['B4', 'B5', 'B6', 'B7'],
                   'C' : ['C4', 'C5', 'C6', 'C7'],
                   'D' : ['D4', 'D5', 'D6', 'D7']})
```

```python
df1
```

```python
df2
```

```python
result = pd.concat([df1, df2])
result
```

```python
df1 = pd.DataFrame({'Student_id': [1,2,3,4],
                   'Name': ['Bruce', 'Clark', 'Steve', 'Diana'],
                   'Age': [40, 33, 98, 356]})

df2 = pd.DataFrame({'Student_id': [2,3,4,5],
                   'Subject': ['Physics', 'Chemistry', 'Maths', 'Biology'],
                   'Marks': [56, 78, 99, 67]})
```

```python
df1
```

```python
df2
```

```python
# Inner Join:
merge_df = pd.merge(df1, df2, on= 'Student_id', how= 'inner')
merge_df
```

```python
# Left Join:
merge_df = pd.merge(df1, df2, on = 'Student_id', how = 'left')
merge_df
```

```python
# Right Join:
merge_df = pd.merge(df1, df2, on = 'Student_id', how = 'right')
merge_df
```

```python
# Full / Outer Join:
merge_df = pd.merge(df1, df2, on = 'Student_id', how = 'outer')
merge_df
```

### loc vs iloc:

The main distinction between the two methods is:

- loc gets rows (and/or columns) with particular labels.

- iloc gets rows (and/or columns) at integer locations.

```python
s = pd.Series(list('abcdef'), index= [49, 48, 47, 0, 1, 2])
s
```

```python
# Value at Index Label 0:
s.loc[0]
```

```python
# Value at Index Location 0:
s.iloc[0]
```

```python
# Rows at Index Lables Between 0 and 1 (Inclusive):
s.loc[0:1]
```

```python
# Rows at Index Location Between 0 and 1 (Exclusive):
s.iloc[0:1]
```

```python
# Getting First two Elements of Series by lables:
s.loc[49:48]
```

```python
# Getting First two Elements of Series by location:
s.iloc[0:2]
```
