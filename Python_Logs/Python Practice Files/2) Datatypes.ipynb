# Datatypes:

Type of Data Stored in a Variable:

#### Numeric:
int : Full Number

float : Number with decimals

complex : Complex Number, With Real and Imaginary Part

#### Boolean:

True (1)

False (0)

#### Sequence:

string : Sequence of Characters

List : Ordered Sequence of Objects (Mutable)

Tuple : Ordered Sequence of Objects (Immutable)

Dictionary : Key-Value Pairs (Unordered)

Set : Unorder Sequence of Unique Objects.

```python
age = 33
height = 1.85
print(type(age), type(height))
```

```python
is_married = False
print(type(is_married))
```

```python
name = 'Bruce'
print(type(name))
```

#### Typecasting:
Changing Type of Data (Variable), by in-built Python Typecasting Functions.

```python
accuracy = 90.1
print(accuracy)
print(type(accuracy))

accuracy_int = int(accuracy)
print(accuracy_int)
print(type(accuracy_int))
```

```python
var = 2
var1 = float(var)
print(var, var1)
print(type(var), type(var1))
```

```python
# Data Lose:
var = 2.9
var1 = int(var)
print(var, var1)
print(type(var), type(var1))
```

```python
var = 2
var1 = str(var)
print(var, var1)
print(type(var), type(var1))
```

```python
var = '2'
var1 = int(var)
print(var, var1)
print(type(var), type(var1))
```

#### String Indexing:

```python
string_var = 'Hello'
print(len(string_var))
print(string_var[0])
print(string_var[1])
print(string_var[2])
print(string_var[3])
print(string_var[4])
```

#### Reverse String Indexing:

```python
string_var = 'Hello'
print(len(string_var))
print(string_var[-1])
print(string_var[-2])
print(string_var[-3])
print(string_var[-4])
print(string_var[-5])
```

```python
# This is not Possible:
string_var[0] = 't'
```

```python
# Re-assignment is Possible:
string_var = 'Hello'
string_var = 'Yellow'
print(string_var)
```
