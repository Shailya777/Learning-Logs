# Tuple:

Ordered Sequence of Objects of Any Data Types.

Immutable

```python
t = (1,2,3,4)
print(t)
print(type(t))
```

```python
t = ('hello', 5, 5.5)
print(t)
print(t[0])
print(t[1])
print(t[2])
```

```python
t = 'hello', 5, 5.5
print(t)
print(type(t))
```

```python
# Length of a Tuple:
print(len(t))
```

```python
# Nested Tuple:
t = (1,2,3,(4,5))
print(len(t))
print(t[0])
print(t[1])
print(t[2])
print(t[3][0])
print(t[3][1])
```

```python
t1 = ('Hello', 5, 5.5, [2,3,4])
print(t1[0])
print(t1[0][0])
print(t1[0][1])
print(t1[0][-1])

print(t1[-1][-1])
print(t1[-1][-2])
print(t1[-1][-3])
```

```python
# Different Ways of Definig Tuples:
t1 = (1,2,3,4)
t2 = 1,2,3,4
t3 = 1,
t4 = (1,)

print(t1)
print(t2)
print(t3)
print(t4)

print(type(t1))
print(type(t2))
print(type(t3))
print(type(t4))

print(len(t1))
print(len(t2))
print(len(t3))
print(len(t4))
```

### Slicing in Tuple:

```python
t = ('Mumbai', 10, 'Python', 100)
print(t)
print(len(t))
```

```python
print(t[0:2])
```

```python
print(t[2:])
```

```python
print(t[:3])
```

```python
print(t[0:4:2])
```

```python
print(t[-2:])
```

```python
print(t[-4:-1])
```

### Tuple Concatenation:

```python
t1 = ('Welcome', 'to', "today's")
t2 = ('class', 'on', 'Python')
t3 = t1 + t2
print(t3)
```

### Functions of Tuples and Lists:

```python
l1 = [1,2,3,4]
print(sum(l1))
print(max(l1))
print(min(l1))
```

```python
t1 = (1,2,3,4)
print(sum(t1))
print(max(t1))
print(min(t1))
```

### Immutability of Tuples:

```python
t = ('Hello!!', 'Welcome', 10, 'world', 'of', 'Analytics!!')
```

```python
print(t[2])
```

```python
t[3] = 'to'
```

### Sorting:

```python
# tuples cannot be Sorted as They are Immutable.
t = (2,3,8,5,4,1)
new_t = sorted(t)
print(t)
print(new_t)
```

```python
# Way to Sort Tuples:
t = (2,3,8,5,4,1)
new_t = sorted(t)
new_t = tuple(new_t)
print(new_t)
```
