# Dictionary

   - Unordered Collection of Key:Value Pair
   - Dictionary stores element as keys and value pairs
   - The key is like an index, its always unique & immutable
   - The values are the objects that contain information
   - Values are accessed using their keys
   - Each key is followed by a value separated by a colon
   - The values can be mutable, & duplicates
   - Each key and value pair is separated by a comma enclosed inside curly brackets

```python
dct = {'Name':'John', 'Age':30}
print(dct)
print(type(dct))
print(len(dct))
```

```python
d1 = {'India':'INR', 'USA':'USD','Hong Kong':'HKD'}
print(d1)
print(type(d1))
print(len(d1))
```

```python
# Accessing values Uisng Keys:
print(d1['India'])
```

```python
# Updating Value for a Key:
d1['Hong Kong'] = 'NEW'
print(d1)
```

```python
# Adding New Key : Value Pair in Dictionary:
d1['Japan'] = 'YEN'
print(d1)
```

```python
# Deleting Key:Value pair from A Dictionary:
del d1['Japan']
print(d1)
```

```python
# Sorting A Dictionary:
print(sorted(d1))
```

```python
# values() Method:
print(d1.values())
```

```python
# keys() Method:
print(d1.keys())
```

```python
# update() Method:
d1.update({'China':'YUAN'})
print(d1)
```

```python
# get() Method:
print(d1.get('China'))
```
