```python
data = {'Year' : [1920, 1930, 1940, 1950, 1960, 1970, 1980, 1990, 2000, 2010, 2020],
       'Exchange Rate' : [65, 69, 71, 64, 62, 59, 71, 71, 75, 78, 81]}
```

```python
print(data)
```

```python
df = pd.DataFrame(data)
df
```

### Graphs Using Pandas Dataframe:

```python
df.plot(x = 'Year', y = 'Exchange Rate', kind = 'bar')
plt.show()
```

```python
df.plot(x = 'Year', y = 'Exchange Rate', kind = 'area')
plt.show()
```

```python
df.plot(x = 'Year', y = 'Exchange Rate', kind = 'barh')
plt.show()
```

```python
df.plot(x = 'Year', y = 'Exchange Rate', kind = 'scatter')
plt.show()
```

### Graphs Using Matplotlib:

```python
plt.bar(df['Year'], df['Exchange Rate'])
plt.show()
```

```python
plt.scatter(x = df['Year'], y = df['Exchange Rate'])
plt.show()
```

```python
df = pd.read_csv('Churn_Modelling.csv')
df.head()
```

```python
df['Geography'].unique()
```

```python
df['Geography'].value_counts()
```

```python
df['Geography'].value_counts().plot()
plt.show()
```

```python
df['Geography'].value_counts().plot(kind = 'bar')
plt.show()
```

```python
df['Geography'].value_counts().plot(kind = 'pie')
plt.show()
```

```python
df['Exited'].value_counts()
```

```python
df['Exited'].value_counts().plot(kind = 'barh')
plt.show()
```

```python
df['EstimatedSalary'].mean()
```

```python
plt.bar(df['Geography'], df['EstimatedSalary'].mean())
plt.title('Geography vs Average Estimated Salary')
plt.show()
```

```python
plt.scatter(x = df['Age'],
           y = df['Tenure'])
plt.title('Scatter Plot')
plt.xlabel('Age')
plt.ylabel('Tenure')
plt.show()
```

```python
df['Tenure'].value_counts()
```

```python
plt.hist(df['Tenure'])
plt.xlabel('Tenure')
plt.show()
```

```python
labels_ex = 'JavaScript', 'Java', 'Python', 'R'
sizes = [15,30,45,10]
explode_labels= (0,0.5,0,0)
fig1, ax1 = plt.subplots()


ax1.pie(sizes, explode = explode_labels, labels=labels_ex, shadow=True, startangle=90)
ax1.axis('equal')
plt.show()
```
