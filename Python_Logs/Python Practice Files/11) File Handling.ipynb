# File Handling:

File handling in Python involves interacting with files on your computer to perform operations such as creating, reading, writing, and appending data.

1. Opening a File:


The open() function is the primary method for file handling. It takes the filename and the mode as arguments:

file_object = open("filename.txt", "mode")

2. Reading:

read(): Reads the entire content of the file.

readline(): Reads a single line from the file.

readlines(): Reads all lines into a list of strings.

3. Writing:

write(string): Writes a string to the file.

writelines(list_of_strings): Writes a list of strings to the file.

4. Closing a File:

After performing operations, it is crucial to close the file using the close() method to release resources and save changes:

file_object.close()

##### Reading A File:

```python
file = open('test_file.txt', 'r')

for line in file:
    print(line)

file.close()
```

```python
file = open('test_file.txt', 'r')
print(file.read())
file.close()
```

```python
file = open('test_file.txt', 'r')
print(file.read(10))
file.close()
```

```python
file = open('test_file.txt', 'r')
print(file.readline())
file.close()
```

```python
file = open('test_file.txt', 'r')
print(file.readlines())
file.close()
```

##### Writing A File:

```python
file = open('test_file.txt', 'w')
file.write('Writing into The File.\n')
file.write('I am Batman!\n')
file.close()
```

```python
file = open('test_file.txt', 'r')
print(file.read())
```

```python
file = open('test_file.txt', 'a')
file.write('We all fly down here.\n')
file.write('No one sneaks up on me!\n')
file.close()
```

```python
file = open('test_file.txt', 'r')
print(file.read())
```

##### Using Context Managers (Recommended):

The with statement provides a more robust way to handle files, ensuring they are automatically closed even if errors occur:
    
    with open("filename.txt", "r") as file_object:
    # Perform operations here
    content = file_object.read()

##### File is automatically closed outside the 'with' block

```python
with open('test_file.txt', 'a') as file:
    file.write('Tell me, Do you Bleed?\n')
    file.write('Why So Serious??\n')

with open('test_file.txt', 'r') as file:
    print(file.read())
```

##### Writing and Reading a file at Same Time:

```python
with open('test_file.txt', 'a+') as file:
    file.write('You just Started a War!\n')
    file.write('I can do this All Day!!\n')

    file.seek(0)
    print(file.read())
```
