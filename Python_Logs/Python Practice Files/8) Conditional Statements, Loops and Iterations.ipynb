# Conditional Statements, Loops and Iterations:

### Conditional Statements:

```python
# Check whether a number is odd or even:
x = 88

if x % 2 == 0:
    print(f'{x} is an even Number.')
else:
    print(f'{x} is an odd Number.')
```

```python
x = int(input('Enter Number: '))

if x % 2 == 0:
    print(f'{x} is an even Number.')
else:
    print(f'{x} is an odd Number.')
```

```python
age = int(input("Enter your age: "))

if age < 18 or age > 50:
    print("Sorry!! You are not allowed for the party")
else:
    print("Welcome to the party")
```

```python
shop_total = int(input("Input your billed amount: "))

if shop_total >= 5000:
    print("You have won a voucher of Rs. 1000/-")
elif shop_total >= 2500:
    print("You have won a voucher of Rs. 500/-")
else:
    print("OOPS!! No discount for you!!")
```

### Iterations:

```python
# Iterate Over a List of Integers:
list1 = [10,20,30,40,50]

for element in list1:
    print(element)
```

```python
# Iterate over a String:
str1 = 'India'

for char in str1:
    print(char)
```

```python
# Iterate over a Dictionary:

student_data = {1: ['Sam', 25], 2: ['Steve', 98], 3: ['Tony', 45], 4: ['Peter', 20], 5: ['Wanda', 35]}

for key, value in student_data.items():
    print(f'Key- {key}: Value- {value}')
```

```python
# Iterating over a Range of Values:
for num in range(1,10):
    print(num)
```

```python
# List Comprehension:
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]
list_2 = []

for element in list_1:
    list_2.append(len(element))

print(list_1)
print(list_2)
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]
list_2 = [len(i) for i in list_1]
print(list_1)
print(list_2)
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]
d1 = {i : len(i) for i in list_1}
print(d1)
```

##### Finding Out All The Prime Numbers from 3-100:

```python
primes = []

for num in range(3,101):
    for i in range(2, num):
        if num % i == 0:
            break
    else:
        primes.append(num)

print(primes)
```

##### Getting a List of All Numbers Divisible by 17 from another List:

```python
list1 = [12, 14, 15, 67, 87, 34, 51, 34, 23, 68, 76, 85, 51, 63, 102, 120, 113, 119, 170, 187, 200]

result = []
for num in list1:
    if num % 17 == 0:
        result.append(num)

print(result)
```

### Loops:

##### Print numbers from 1 to 10:

```python
num = 1
while num <= 10:
    print(num)
    num += 1
```

##### Find sum of first 5 natural numbers:

```python
n = 5
total = 0
i = 1
while i <= n:
    total += i
    i += 1
print("Sum =", total)
```

##### Reverse the digits of a number:

```python
num = 1234
rev = 0
while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10
print("Reversed number =", rev)
```

##### Keep asking user until they enter 'yes':

```python
answer = ""
while answer.lower() != "yes":
    answer = input("Type 'yes' to continue: ")
print("Thank you!")
```

##### Print first 10 Fibonacci numbers:

```python
a, b = 0, 1
count = 0
while count < 10:
    print(a, end=" ")
    a, b = b, a + b
    count += 1
```
