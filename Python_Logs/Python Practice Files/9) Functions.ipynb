##### Function to Find Whether a Number id Even or Odd:

```python
def even_odd(num):
    if num % 2 == 0:
        print(f'{num} is Even.')
    else:
        print(f'{num} is Odd.')
```

```python
even_odd(14)
even_odd(567)
even_odd(16789)
```

##### Function To Find Factorial of a Number:

```python
def factorial(num):
    fact = 1
    
    for i in range(1, num +1):
        fact = fact * i
    
    return fact
```

```python
print(factorial(5))
print(factorial(10))
print(factorial(1))
print(factorial(3))
print(factorial(4))
```

##### Sum of All Natuaral Numbers till Number Provided:

```python
def sum_n(num):
    total = 0
    
    for i in range(1, num+1):
        total += i

    return total
```

```python
print(sum_n(2))
print(sum_n(3))
print(sum_n(4))
print(sum_n(5))
print(sum_n(10))
```

##### Function to Find if Given Number or String is Palindrome or not (Reads Same from Both Sides):

```python
def is_palindrome(arg):
    arg = str(arg)
    return arg == arg[::-1]
```

```python
print(is_palindrome('madam'))
print(is_palindrome(12321))
print(is_palindrome('Hello'))
print(is_palindrome(12323))
```

##### Function With Default Values:

```python
def generic_func(arg1 = 0, arg2 = 0):
    return arg1 + arg2
```

```python
print(generic_func())
print(generic_func(10))
print(generic_func(12,10))
```

### Lambda Functions:

- A lambda function in Python is an anonymous, single-expression function defined using the lambda keyword.

-  Unlike regular functions defined with def, lambda functions do not require a name and are typically used for short, one-time operations.

- Anonymous:
Lambda functions do not have a defined name unless explicitly assigned to a variable.

- Single Expression:
They are restricted to a single expression, which is automatically evaluated and returned. An explicit return statement is not needed.

- Concise:
Their single-line syntax makes them suitable for quick, inline function definitions.

- Arguments:
They can accept any number of arguments, similar to regular functions, including positional, named, variable positional, and variable keyword arguments.

##### Lambda function that adds 10 to the number passed as an arguement, and print the result:

```python
x = lambda a : a + 10
```

```python
print(x(10))
print(x(100))
print(x(1000))
print(x(10000))
```

##### #### Create a lambda function that checks a number is even or odd:

```python
even_odd = lambda num : 'Even' if num % 2 == 0 else 'Odd'
```

```python
print(even_odd(10))
print(even_odd(11))
print(even_odd(12))
print(even_odd(134))
print(even_odd(547))
```

##### Lambda function that adds two integers:

```python
add = lambda n1, n2 : n1 + n2
```

```python
print(add(5,5))
print(add(5,15))
print(add(15,45))
print(add(35,12))
print(add(31,47))
```
