```python
# Importing Seaborn:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Loading a Dataset from Seaborn in Dataframe:
df = sns.load_dataset('iris')
df.head()
```

```python
print(df.shape)
```

```python
df.describe()
```

```python
df.info()
```

```python
df['species'].value_counts()
```

```python
sns.kdeplot(data = df['sepal_length'], fill= True)
plt.show()
```

```python
print(df.columns)
```

```python
for column in ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']:
    sns.kdeplot(df[column], fill= True)
```

```python
# DistPlot:
sns.displot(df['sepal_length'])
plt.show()
```

```python
# PairPlot:
sns.pairplot(df, hue= 'species')
plt.show()
```

```python
data = {'Year': [1920,1930,1940,1950,1960,1970,1980,1990,2000, 2010, 2020],
       'Exchange Rate': [65, 69, 71, 64, 62, 59, 72, 71, 75, 78, 81]}
data
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
sns.lineplot(x = df['Year'], y = df['Exchange Rate'])
plt.show()
```

```python
# Load the example flights dataset and convert to long-form
flights_long = sns.load_dataset("flights")
flights = flights_long.pivot(index="month", columns="year", values="passengers")

flights.head(5)
```

```python
# Heatmap:
plt.figure(figsize = (15,12))
sns.heatmap(data= flights, annot= True, fmt = 'd')
plt.show()
```

```python
# Boxplot:
sns.boxplot(y = df['Exchange Rate'])
plt.show()
```
