### List Concatenation:

```python
# String Concat:
a = 'Hello'
b = 'World!'
print(a + b)
```

```python
# List Concat:
list1 = [1,2,3,4]
list2 = [5,6,7,8]
new_list = list1 + list2
print(new_list)
```

### Membership in Lists:

```python
list3 = [1,2,3,4]
print(1 in list3)
print(5 in list3)
```

### List Mutability:

```python
list1 = ['Maths', 'Physics', 'Chemistry', 'Biology']
print(list1)
```

```python
print(list1[3])
```

```python
list1[3] = 'Computer Science'
print(list1)
```

```python
# Extend: Extends The List:
list1 = [1,2,3,4]
list1.extend([5,6])
print(list1)
```

```python
# Append: Appends to Elements of List:
list1 = [1,2,3,4]
list1.append([5,6])
print(list1)
```

```python
# Deletes Element from list:
list1 = [1,2,3,4]
del list1[2]
print(list1)
```

```python
# Pop: Deletes the Object Based on The Given Index Value
# Default: Removes Lase Element.

list1 = [11,22,33,44]
print(list1.pop())
print(list1.pop(0))
print(list1)
```

```python
# Remove: Removes Element Based on Given Value.
list1 = [1,2,3,4]
list1.remove(4)
print(list1)
```

```python
# Sorting:
a = [9,2,5,1,17,23,-1]
a.sort() # Ascending Order
print(a)
```

```python
a.sort(reverse= True) # Descending
print(a)
```

### Difference Between Sort and Sorted:

```python
A = ['Orange', 'Strawberry', 'Mango']
B = A.sort()
print(A)
print(B)
```

```python
A = ['Orange', 'Strawberry', 'Mango']
B = sorted(A)
print(A)
print(B)
```

### Shallow and Deep Copy:

```python
# Shallow Copy:
A = ['Orange', 'Strawberry', 'Mango']
B = A
print(A)
print(B)
```

```python
print(A.pop())
```

```python
print(A)
print(B)
```

```python
A = ['Orange', 'Strawberry', 'Mango']
B = A[0:3]
print(A)
print(B)
```

```python
print(A.pop())
```

```python
print(A)
print(B)
```
