# Arithmetic Operations:

```python
# Addition:
a = 10
b = 12
c = a + b
print(a, b, c)
```

```python
# Subtraction:
a = 10
b = 12
c = a - b
print(a, b, c)
```

```python
# Multiplication:
a = 10
b = 12
c = a * b
print(a, b, c)
```

```python
# Division:
a = 10
b = 12
c = a / b
print(a, b, c)
```

```python
# Floor Division:
a = 100
b = 12
c = a // b
print(a, b, c)
```

```python
# Modulus / Remainder:
a = 100
b = 12
c = a % b
print(a, b, c)
```

```python
# Exponent (Power):
a = 2
b = 5
c = a ** b
print(a, b, c)
```

### Order of Precedence:

PEMDAS : 
Parentheses : ()
Exponents : **
Division , Multiplication : / , * , // , %
Addition, Subtraction : + , -

```python
print((4 * 5) - 9 + 6 / 7)
```

```python
print(4 % (1 + 9)**2 - 60 // (7 + 2))
```
