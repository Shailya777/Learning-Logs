# 6. n-Grams

### What are n-Grams?
An **n-gram** is a contiguous sequence of *n* items (words) from a given sample of text.

* **Unigram (n=1):** "Data", "Science"
* **Bigram (n=2):** "Data Science", "New York"
* **Trigram (n=3):** "Natural Language Processing"

### Why use them?
Standard Bag-of-Words loses context because it looks at words individually. n-Grams capture **local context**. 
* "New" and "York" separately are just adjectives and nouns. "New York" together is a specific location.

### Applications
* **Next Word Prediction:** (e.g., auto-complete on your phone).
* **Plagiarism Detection:** Matching sequences of words.

```python
from nltk import ngrams
from sklearn.feature_extraction.text import CountVectorizer

text = 'The quick brown fox jumps'
token = text.split()

# NLTK:
print('NLTK Bigrams:')
bigrams = list(ngrams(token, 2))
print(bigrams)

print('NLTK Trigrams:')
trigrams = list(ngrams(token, 3))
print(trigrams)

# SKLEARN:
vectorizer = CountVectorizer(ngram_range= (2,2)) # ngram_range=(2, 2) means only bigrams
X = vectorizer.fit_transform([text])

print('Sklearn Bigrams:')
print(vectorizer.get_feature_names_out())
```
