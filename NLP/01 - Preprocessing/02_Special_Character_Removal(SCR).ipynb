# 2. NLP Preprocessing: Special Character Removal (SCR)

### What is SCR?
Special Character Removal involves stripping non-alphanumeric characters like punctuation (`!`, `?`, `.`), symbols (`@`, `#`, `$`), and sometimes emojis from the text.

### Why do we use it?
* **Noise Reduction:** Punctuation often doesn't add semantic meaning for tasks like topic modeling or spam classification.
* **Tokenization Safety:** Some simple tokenizers might accidentally attach punctuation to words (e.g., treating "word." as a different token than "word").

### 🛠️ Common Tools
* **Regex (Regular Expressions):** The most flexible way to find and replace patterns.
* **String Library:** Python's built-in list of punctuation characters.

```python
import re
import string

text = 'Hello World!!! How are you doing? #NLP #Learning @2026 :)'

# Method 1: Using string.punctuation
translator = str.maketrans('', '', string.punctuation) # string.punctuation contains: !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
clean_text = text.translate(translator)

print(f'Original: {text}')
print(f'Cleaned text: {clean_text}')

# Method 2: Using Regex (Good for specific cleaning, like keeping # or @)
clean_text = re.sub(r'[^a-zA-Z0-9\s]', '', text) # The pattern [^a-zA-Z0-9\s] means: Remove anything that is NOT a letter, number, or space

print('\n')
print(f'Original: {text}')
print(f'Cleaned text: {clean_text}')
```
