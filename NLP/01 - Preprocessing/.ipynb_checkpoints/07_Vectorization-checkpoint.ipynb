# 7. Vectorization (BoW & TF-IDF)

### What is Vectorization?
Machine Learning models cannot read text; they only understand numbers. Vectorization is the process of converting text into numerical vectors.

### 1. Bag of Words (BoW)
* Creates a matrix where rows are documents and columns are unique words.
* The value is the **Count** (frequency) of that word.
* **Issue:** Common words like "the" dominate the counts.

### 2. TF-IDF (Term Frequency - Inverse Document Frequency)
* Rescales the counts to highlight "interesting" words.
* **TF:** How often word appears in *this* document.
* **IDF:** Downweights words that appear in *every* document (like "the", "is").
* **Result:** Rare words that are specific to the current document get high scores.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

corpus = [
    'I love deep learning.',
    'I love NLP.',
    'Deep learning is the future.'
]

# 1: Bag of Words (CountVectorizer):
cv = CountVectorizer()
X_bow = cv.fit_transform(corpus)

print('Bag of Words Matrix:')
df_bow = pd.DataFrame(X_bow.toarray(),
                     columns= cv.get_feature_names_out())
print(df_bow)
print('\n')

# 2: TF_IDF:
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(corpus)

print('TF-IDF Matrix:')
df_tfidf = pd.DataFrame(X_tfidf.toarray(),
                       columns= tfidf.get_feature_names_out())
print(df_tfidf)
```
