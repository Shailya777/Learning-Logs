# 5. Stop Word Removal

### What are Stop Words?
Stop words are high-frequency words that carry very little semantic meaning in a sentence. Examples include: *the, is, at, which, on*.

### Why Remove Them?
* **Noise Reduction:** Allows the model to focus on the "meat" of the sentence (nouns, verbs, adjectives).
* **Dimensionality Reduction:** Reduces the size of the vocabulary in Bag of Words models.

### ⚠️ When to Keep Them?
* **Machine Translation:** "The" vs "A" makes a difference.
* **Sentiment Analysis:** "Not" is a stop word, but removing it flips the meaning of "Not good."

```python
import nltk
from nltk.corpus import stopwords
import spacy
```

```python
nltk.download('stopwords')
nlp = spacy.load('en_core_web_sm')
```

```python
text = 'This is a sample sentence, showing off the stop words filtration.'
```

```python
# Stop Words Removal using NLTK:
stop_words_nltk = set(stopwords.words('english'))

tokens = nltk.word_tokenize(text)
filtered_nltk = [w for w in tokens if w.lower() not in stop_words_nltk]

print(f'Original: {text}')
print(f'NLTK Filtered: {filtered_nltk}')
```

```python
# Stop Words Removal using SPACY:
doc = nlp(text)
filtered_spacy = [token.text for token in doc if not token.is_stop]

print(f'Original: {text}')
print(f'SPACY Filtered: {filtered_spacy}')
```
