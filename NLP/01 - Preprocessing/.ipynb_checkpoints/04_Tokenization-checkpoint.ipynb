# 4. Tokenization

### What is Tokenization?
Tokenization is the process of breaking down a stream of text into smaller units called **tokens**. These tokens can be words, characters, or sub-words. It is usually the very first step in any NLP pipeline.

### Types of Tokenization
1. **Word Tokenization:** Splitting sentences into individual words.
   * *Example:* "I love AI" $\rightarrow$ `['I', 'love', 'AI']`
2. **Sentence Tokenization:** Splitting a paragraph into sentences.
   * *Example:* "Hello. How are you?" $\rightarrow$ `['Hello.', 'How are you?']`

### Libraries
* **NLTK:** Great for explicit control and academic use.
* **spaCy:** Fast, robust, and handles complex cases (like "U.S.") automatically.

```python
import nltk
import spacy
```

```python
# Downloading nltk Data:
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
import spacy.cli
spacy.cli.download('en_core_web_sm')
```

```python
# Tokenization Using NLTK:

text = 'Dr. Strange loves AI. He lives in the U.S. and pays $500.'

from nltk.tokenize import word_tokenize, sent_tokenize

print('NLTK Tokenization:')
print(f'Sentences: {sent_tokenize(text)}')
print(f'Words: {word_tokenize(text)}')
```

```python
# Tokenization using Spacy:
text = 'Dr. Strange loves AI. He lives in the U.S. and pays $500.'

nlp = spacy.load('en_core_web_sm')
doc = nlp(text)

print('Spacy Tokenization:')
for token in doc:
    print(token.text)
```
