# 1. NLP Preprocessing: Case Folding

### What is Case Folding?
Case folding is the process of converting all text to the same casing, typically **lowercase**. In the eyes of a computer, "Apple" and "apple" are two completely different strings. Case folding unifies them.

### Why do we use it?
* **Vocabulary Reduction:** It significantly shrinks the size of the vocabulary (features) the model has to learn.
* **Consistency:** Ensures that words at the beginning of sentences (capitalized) are treated the same as words in the middle.

### ⚠️ When NOT to use it?
* **Sentiment Analysis:** sometimes ALL CAPS denotes anger or excitement.
* **Named Entity Recognition (NER):** Capitalization is a key feature to identify Proper Nouns (e.g., "US" the country vs. "us" the pronoun).

### Example
* **Input:** "The Quick Brown Fox"
* **Output:** "the quick brown fox"

```python
# Standard Python Approcah:
text_1 = 'The Quick Brown FOX jumps over the Lazy Dog.'

lower_text = text_1.lower()

print(f'Original: {text_1}')
print(f'Lower Cased: {lower_text}')
```

```python
# Verifying that 'Fox' and 'fox' are same:
print(f'Is "Fox" == "fox"?: {"Fox" == "fox"}')
print(f'Is "Fox".lower() == "fox"?: {"Fox".lower() == "fox"}')
```
