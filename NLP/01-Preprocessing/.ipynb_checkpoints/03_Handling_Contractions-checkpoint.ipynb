# 3. NLP Preprocessing: Handling Contractions

### What are Contractions?
Contractions are shortened versions of words or syllables using an apostrophe (e.g., "don't" for "do not", "I've" for "I have").

### The Problem in NLP
Standard tokenizers often split contractions incorrectly:
* "don't" might become `["don", "'", "t"]` or `["don", "t"]`.
* This confuses the model because "don" is not a standard word, and "t" loses the meaning of "not".

### The Solution: Expansion
We replace contractions with their full forms **before** tokenization.
* `don't` $\rightarrow$ `do not`
* `I'll` $\rightarrow$ `I will`

### Library
The `contractions` library is the standard tool for this.

```python
# Run following code if contractions library is not already insalled:

#!pip install contractions
```

```python
import contractions

text = "I've decided that I won't give up on NLP. It's tough but it'll be worth it."

# Expanding Contractions:
expanded_text = []
for word in text.split():
    expanded_text.append(contractions.fix(word)) # fix method expands the word

# Joining back into a string:
final_text = ' '.join(expanded_text)

print(f'Original: {text}')
print(f'Expanded Text: {final_text}')
```
