# Word2Vec: CBOW and Skip-Gram

## 1. The Word2Vec Revolution
Created by Tomas Mikolov at Google (2013), Word2Vec revolutionized NLP. It is based on the **Distributional Hypothesis**: *"You shall know a word by the company it keeps."*

It is a shallow neural network (just 1 hidden layer) trained to reconstruct linguistic contexts of words.

## 2. Architecture 1: CBOW (Continuous Bag of Words)
* **Goal:** Predict the **Target Word** (middle word) based on the **Context Words** (surrounding words).
* **Input:** "The", "quick", "fox", "jumps"
* **Target:** "brown"
* **Pros:** Faster to train; better accuracy for frequent words.
* **Cons:** Smoothes over distributional information; struggles with rare words.

## 3. Architecture 2: Skip-Gram
* **Goal:** Predict the **Context Words** given a single **Target Word**.
* **Input:** "brown"
* **Target:** "The", "quick", "fox", "jumps"
* **Pros:** Works well with small amounts of training data; represents rare words or phrases well.
* **Cons:** Slower to train.

## 4. Key Parameters
* **Vector Size:** Dimension of the embedding (usually 100-300).
* **Window:** How many neighbors to look at (e.g., 5 words to the left/right).
* **Min Count:** Ignore words that appear less than $X$ times.
