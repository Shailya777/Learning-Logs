# Advanced Embeddings: GloVe & FastText

## Part 1: GloVe (Global Vectors)
Developed by Stanford. Unlike Word2Vec (which learns locally, window by window), GloVe learns by constructing a massive **Co-occurrence Matrix** of the entire corpus first.
* It counts how often word $i$ appears in the context of word $j$ across the *whole* dataset.
* It factorizes this matrix to produce vectors.
* **Pros:** Captures global statistics better.
* **Cons:** Hard to train online (cannot easily add new words without rebuilding the matrix).

## Part 2: FastText
Developed by Facebook (Meta). It improves upon Word2Vec by treating words as bags of **Character n-grams**.
* The word "apple" is broken down into `<ap, app, ppl, ple, le>`, plus the whole word `<apple>`.
* **The Killer Feature:** It can generate embeddings for **Out-Of-Vocabulary (OOV)** words!
    * If you train on "apple" but test on "apple-sauce", Word2Vec fails. FastText looks at the n-grams "apple" and "sauce" and constructs a vector.

## Part 3: Comparison
| Feature | Word2Vec | GloVe | FastText |
| :--- | :--- | :--- | :--- |
| **Training** | Prediction (Local) | Count (Global) | Prediction (Sub-word) |
| **Speed** | Fast | Moderate | Slowest (more computation) |
| **OOV Words** | Fails | Fails | **Handles them!** |
