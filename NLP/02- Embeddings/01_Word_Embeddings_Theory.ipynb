# Word Embeddings: The Foundation

## 1. What are Word Embeddings?
In the Bag-of-Words (BoW) model, we represented words as sparse vectors of counts (0s and 1s). This had two huge problems:
1. **Sparsity:** A vocabulary of 100,000 words meant a vector size of 100,000 with mostly zeros.
2. **No Meaning:** "King" and "Prince" were just as different as "King" and "Banana". The computer didn't know they were related.

**Word Embeddings** solve this by representing words as **Dense Vectors** (lists of floating-point numbers) in a continuous vector space.
* instead of `[0, 0, 1, 0...]` (size 100k)
* we get `[0.25, -0.4, 0.11, ...]` (size 50 or 300)

## 2. Why use them?
The magic of embeddings is **Semantic Relationship**. Words that are used in similar contexts will have vectors that are mathematically close to each other.
* The distance between *King* and *Queen* is roughly the same as *Man* and *Woman*.
* This allows for vector arithmetic: `King - Man + Woman = Queen`.

## 3. Types of Embeddings
1. **Prediction-Based (Word2Vec, FastText):** Uses a neural network to predict a word given its neighbors.
2. **Frequency-Based (GloVe):** Uses a co-occurrence matrix to count how often words appear together globally.
3. **Contextual (BERT, ELMo):** (Modern) The embedding for "bank" changes depending on if you mean "river bank" or "finance bank".

## 4. Most Common Libraries
* **Gensim:** The absolute industry standard for training Word2Vec, FastText, and loading GloVe.
* **spaCy:** Comes with high-quality pre-trained vectors attached to its models.
* **TensorFlow / PyTorch:** Used if you want to build an Embedding Layer from scratch for a specific Deep Learning task.
