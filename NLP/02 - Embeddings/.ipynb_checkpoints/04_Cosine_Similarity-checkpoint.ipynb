# Cosine Similarity

## 1. The Math of Similarity
When we have word vectors, how do we know if two words are similar? We cannot use standard distance (Euclidean) because the *magnitude* (length) of the vector often depends on word frequency, not meaning.

Instead, we look at the **Angle** between the vectors.

## 2. The Formula
$$\text{Cosine Similarity} = \frac{A \cdot B}{||A|| \times ||B||}$$

* **1.0:** Vectors point in exactly the same direction (Synonyms).
* **0.0:** Vectors are orthogonal (90 degrees). Unrelated.
* **-1.0:** Vectors are opposite (180 degrees). Antonyms (conceptually).

In high-dimensional space (100d+), most random vectors have a similarity near 0.

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Manually creating 3 vectors, 2 are mostly similar, 3rd is very different:
vec_king = np.array([[5.1, 3.2, 1.5]])
vec_queen = np.array([[5.0, 3.3, 1.6]])
vec_apple = np.array([[0.1, 0.2, 8.5]])
```

```python
# Calculating Cosine Similarity:
sim_king_queen = cosine_similarity(vec_king, vec_queen)
sim_king_apple = cosine_similarity(vec_king, vec_apple)
```

```python
print(f'Similarity (King vs Queen): {sim_king_queen}')
```

```python
print(f'Similarity (King vs Apple): {sim_king_apple}')
```
