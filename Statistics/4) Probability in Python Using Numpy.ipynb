### We can Simulate Flipping a Coin Using Random Number Generation:

```python
coin = np.random.randint(0,2)

if coin == 1:
    print('Head')
else:
    print('Tail')
```

### Simulating Flipping a Coin 1000 times and Counting Heads and Tails:

```python
coin = np.random.randint(0, 2, 1000)

heads = sum(coin)

print(f'Number of Heads: {heads}')
print(f'Number of Tails: {1000 - heads}')
```

### Rolling a Die 6000 Times and Plotting Occurance of Each Number:

```python
dice = np.random.randint(1, 7, 6000)
sns.countplot(x = dice)
plt.show()
```

```python
# While Rolling a Die, Probability of Each Number Ocuuring is Always 1/6.

# Here, In the Graph Above, we can see that,
# after Rolling a Die 6000 times,
#Each Number has Occurred Nearly 1000/6000 Times.
```

### if Two Dice are Rolled, What is the Probability of Getting The Sum of Numbers Greater Than 7?

Answer: 15/36 = 0.4166

##### Simulating Rolling 2 Dice 10 Times:

```python
die1 = np.random.randint(1, 7, 10)
die2 = np.random.randint(1, 7, 10)
```

```python
sum_dice = die1 + die2
print(sum_dice)
```

##### Simulating Rolling 2 Dice 100000 Times and Counting Occurances of Sum Being more than 7:

```python
die1 = np.random.randint(1, 7, 100000)
die2 = np.random.randint(1, 7, 100000)

sum_dice = die1 + die2
```

```python
print(sum(sum_dice > 7))
```

```python
# As we can see, after Rolling Two Dice 100000 Times,
# Occurance of Sum of Two Dice more than 7 is close to Theoritical 
# Probability of 0.4166.
```
