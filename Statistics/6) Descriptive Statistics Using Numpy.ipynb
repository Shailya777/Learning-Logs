# Descriptive Statistics Using Numpy

```python
import numpy as np
```

#### Descriptive Statistics of List using Numpy:

```python
lst = [12, 15, 17, 90, 31, 42, 54, 32, 12, 12]
```

```python
# Mean:

np.mean(lst)
```

```python
# Median:

np.median(lst)
```

```python
# Standard Deviation:

np.std(lst)
```

```python
# Variance:

np.var(lst)
```

#### Descriptive Statistics of 1-D Array using Numpy:

```python
array_1d = np.random.randint(0, 50, 25)
```

```python
print(array_1d)
```

```python
# Mean:
print(np.mean(array_1d))
print(array_1d.mean())
```

```python
# Median:
print(np.median(array_1d))
```

```python
# Variance:
print(np.var(array_1d))
print(array_1d.var())
```

```python
# Standard Deviation:
print(np.std(array_1d))
print(array_1d.std())
```

#### Descriptive Statistics of 2-D Array using Numpy:

```python
array_2d = array_1d.reshape(5,5)
print(array_2d)
```

```python
# Mean of All Elements:
print(np.mean(array_2d))
print(array_2d.mean())
```

```python
# Median of All Elements:
print(np.median(array_2d))
```

```python
# Variance of All Elements:
print(np.var(array_2d))
print(array_2d.var())
```

```python
# Standard Deviation of All Elements:
print(np.std(array_2d))
print(array_2d.std())
```

##### Descriptive Stats Across Rows / Columns of 2-D Array:
    axis = 0 : Across The Rows
    axis = 1 : Across The Columns

```python
print(array_2d)
```

```python
# Mean Across Rows:
print(np.mean(array_2d, axis= 0))
print(array_2d.mean(axis= 0))
```

```python
# Mean Across Columns:
print(np.mean(array_2d, axis= 1))
print(array_2d.mean(axis= 1))
```

```python
# Median Acroass Rows:
print(np.median(array_2d, axis= 0))
```

```python
# Median Across Columns:
print(np.median(array_2d, axis= 1))
```

```python
# Variance Across Rows:
print(np.var(array_2d, axis= 0))
print(array_2d.var(axis= 0))
```

```python
# Variance Across Columns:
print(np.var(array_2d, axis= 1))
print(array_2d.var(axis= 1))
```

```python
# Standard Deviation Across Rows:
print(np.std(array_2d, axis= 0))
print(array_2d.std(axis= 0))
```

```python
# Standard Deviation Across Columns:
print(np.std(array_2d, axis= 1))
print(array_2d.std(axis= 1))
```
