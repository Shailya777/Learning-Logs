## Central Limit Theorem Using Python:

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Central Limit Theorem Using One Population:

##### Rolling a Die 1000000 Times and Plotting:

```python
die_million = np.random.randint(1, 7, 1000000)
print(die_million)
```

```python
sns.displot(x= die_million, bins= 6)
plt.show()
```

As we can see, It is a Uniform Distribution.

```python
# Mean of the Population:
np.mean(die_million)
```

```python
# Standard Deviation of The Population:
np.std(die_million)
```

##### Drawing 250000 Samples of Size 4 from Population:

```python
samples_4 = die_million.reshape(250000, 4)
print(samples_4)
```

```python
# Mean of These Samples:
mean_4 = samples_4.mean(axis= 1) # Mean Across The Columns
print(mean_4)
```

```python
sns.displot(mean_4, bins= 6)
plt.show()
```

As we can see, Distribution of Sample Means Closely Resembles Normal Distribution.

```python
# Mean of Sample Means:
np.mean(mean_4)
```

```python
# Standard Deviation of Sample Means:
np.std(mean_4)
```

Mean of the Sample Means is Close to The Population Mean.

Standard Deviation of Sample Means is Close to value of Population Mean / Square Root of Sample Size. (1.7063196119364625 / 2).

#### Central Limit Theorem Using Two Populations:

```python
# np.random.norma(mean, std, size)
var1 = np.random.normal(loc= 100, scale= 5, size= 10000)
print(var1)
```

```python
# np.random.norma(mean, std, size)
var2 = np.random.normal(loc= 80, scale= 5, size= 10000)
print(var2)
```

##### Mixing Two Populations and Shuffling The Data:

```python
comb_data = np.append(var1, var2)
print(comb_data)
```

```python
np.random.shuffle(comb_data)
```

```python
print(comb_data)
```

```python
sns.displot(comb_data)
plt.show()
```

```python
# Mean of the Mixed Data:
np.mean(comb_data)
```

```python
# Standard Deviation of Mixed Data:
np.std(comb_data)
```

##### Drawing 10000 Samples of Size 2 from Mixed Data:

```python
sample_2 = comb_data.reshape(10000, 2)
print(sample_2)
```

```python
# Means of The Samples:
mean_2 = sample_2.mean(axis= 1)
print(mean_2)
```

```python
sns.displot(mean_2)
plt.show()
```

```python
# Mean of The Sample Means:
np.mean(mean_2)
```

```python
# Standard Deviation of Sample Means:
np.std(mean_2)
```

Mean of the Sample Means is Close to The Population Mean.

Standard Deviation of Sample Means is Close to value of Population Mean / Square Root of Sample Size. (11.169448196122632 / 2**0.5).
