## Assignment 2 - Pie Chart:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
cause_of_death = ["Chronic Diseases", "Uninentional Injuries", "Alzheimers", "Influenza and Pneumonia", "Sepsis", "Others"]
```

```python
percentile = [62,5,4,2,1,26]
```

```python
plt.figure(figsize= (15,12), dpi= 200)

ex = (0.05,0,0,0,0,0.02)

plt.pie(percentile, labels= cause_of_death, explode= ex, autopct="%1.1f%%", startangle= 70)

plt.title("Leading Causes of Death", fontsize= 20)
plt.show()
```
