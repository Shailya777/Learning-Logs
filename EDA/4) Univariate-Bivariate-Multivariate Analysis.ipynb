# Uni-Variate, Bi-Variate, Multi-Variate Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Churn_Modelling.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

### 1) Uni-Variate Analysis:

##### 1) Numerical Variable:

```python
# Distribution of Age:

plt.figure(figsize= (6,4))
sns.histplot(data= df['Age'], bins= 30, color= 'skyblue')
plt.title('Distribution of Age')
plt.show()
```

Interpretation: 

The age distribution is right-skewed. Most customers are between 30–40 years old, with fewer in older age groups.

```python
# Boxplot of Balance:

plt.figure(figsize= (6,4))
sns.boxplot(data= df, x= 'Balance', color= 'orange')
plt.title('Boxplot of Balance')
plt.show()
```

Interpretation:

Many customers have zero balance, while others show a wide range. The long whiskers indicate large variations in account balances.

##### 2) Categorical Variable:

```python
# Countplot of Gender:

plt.figure(figsize= (6,4))
sns.countplot(data= df, x= 'Gender', palette= 'Set2', hue= 'Gender')
plt.title('Count of Gender')
plt.show()
```

Interpretation:

The dataset has a slightly higher proportion of male customers compared to female customers.

```python
# Countplot of Geography:

plt.figure(figsize= (6,4))
sns.countplot(data= df, x= 'Geography', hue= 'Geography', palette= "Set3")
plt.title('Count of Customers by Geography')
plt.show()
```

Interpretation:

Most customers are from France, followed by Spain and Germany.

### 2) Bi-Variate Analysis:

##### 1) Numerical vs Numerical:

```python
# Age vs Balance:

plt.figure(figsize= (6,4))
sns.scatterplot(data= df, x= 'Age', y= 'Balance', alpha= 0.5)
plt.title("Age vs Balance")
plt.show()
```

Interpretation:

No strong relationship is visible between Age and Balance. Balance seems independent of age groups.

##### 2) Numerical vs Categorical:

```python
# Age vs Exited:

plt.figure(figsize= (6,4))
sns.boxplot(data= df, x= 'Exited', y= 'Age', palette= 'coolwarm', hue= 'Exited')
plt.title('Age Distribution by Exit Status')
plt.show()
```

Interpretation:

Customers who exited tend to be slightly older on average than those who stayed.

##### 3) Categorical vs Categorical:

```python
# Gender vs Existed:

plt.figure(figsize= (6,4))
sns.countplot(data= df, x= 'Gender', hue= 'Exited', palette= 'Set1')
plt.title('Churn by Gender')
plt.show()
```

Interpretation:

Both genders show churn, but the churn rate seems slightly higher for females.

```python
# Geography vs Exited:

plt.figure(figsize= (6,4))
sns.countplot(data=df, x= 'Geography', hue= 'Exited', palette= 'Set2')
plt.title('Churn by Geography')
plt.show()
```

Interpretation:

Customers from Germany have a noticeably higher churn rate compared to France and Spain.

### 3) Multi-Variate Analysis:

##### 1) Pairplot for Nymerical Variables:

```python
sns.pairplot(df[['CreditScore', 'Age', 'Balance', 'EstimatedSalary', 'Exited']], 
             hue='Exited', palette='husl')
plt.show()
```

Interpretation:

Pairwise plots show some separation for churners by Age and Balance. Other variables like Salary appear less related to churn.

##### 2) Heatmap of Correlation Matrix:

```python
plt.figure(figsize= (8,4))
sns.heatmap(df[['CreditScore','Age','Tenure','Balance','NumOfProducts','EstimatedSalary','Exited']].corr(), 
            annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

Interpretation:

Most variables are weakly correlated with each other. 'Exited' shows slight positive correlation with Age and negative correlation with NumOfProducts.

##### 3) Age vs Balance vs Exited:

```python
plt.figure(figsize=(6,4))
sns.scatterplot(data= df, x= 'Age', y= 'Balance', hue= 'Exited', alpha=0.5, palette= 'Set1')
plt.title("Age vs Balance Colored by Churn")
plt.show()
```

Interpretation:

Customers in the 40–60 age group with mid-to-high balances show higher churn rates compared to younger customers.

### Summary:

#### **Univariate Analysis**
- **Age:** Most customers are between 30–40 years old, with churn being more common among older customers.  
- **Balance:** Many customers have zero balance, while others vary widely — indicating diverse financial behavior.  
- **Gender & Geography:** Slightly more male customers. Most customers are from France, followed by Spain and Germany.  

#### **Bivariate Analysis**
- **Age vs Churn:** Older customers tend to leave more often than younger customers.  
- **Gender vs Churn:** Females show a slightly higher churn rate than males.  
- **Geography vs Churn:** German customers have the highest churn rate, while French customers are most likely to stay.  
- **Balance vs Age:** No clear relationship, showing churn is not simply dependent on wealth or age alone.  

#### **Multivariate Analysis**
- **Correlation Heatmap:** No strong linear correlations. Churn shows weak relationships with **Age** (positive) and **NumOfProducts** (negative).  
- **Scatterplots:** Middle-aged customers (40–60) with higher balances tend to churn more.  
- **Pairplots:** Visual separation between churners and non-churners is visible mostly in Age and Balance dimensions.  

---

#### **Key Takeaways**
- **Target Segment for Churn Prevention:** Middle-aged German customers, especially those with higher balances, need more engagement.  
- **Less Influential Factors:** Estimated Salary and Credit Score show little impact on churn in this dataset.
