## Types of Plots

### Different Types of Plots can be created using Matplotlib:

### 1) Histogram:

#### Graphical Representation of a Probability Distribution.
#### Can be created using hist function from Matplotlib.

#### Advantages of Histogram:
#### 1) Displays Number of Values within a specified Interval.
#### 2) Suitable for Large Datasets as they can be grouped within the Intervals.

### 2) Scatter Plot:

#### Used to graphically display Relationship Between Variables.
#### Can be created using plot or scatter method.

#### Advantages of Scatter Plot:
#### 1) Shows Correlation Between variables.
#### 2) Suitable for large Datasets.
#### 3) Easy to find Clusters.
#### 4) Possible to represent each point of Data as a Point on the Plot.

### Demo: Histogram and Scatter Plot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import style
```

```python
boston_df = pd.read_csv("BostonHouse.csv")
```

```python
boston_df.head()
```

```python
boston_df.shape
```

```python
boston_df.info()
```

```python
# Histogram of Prices of House:
```

```python
plt.figure(figsize= (15,8))

plt.hist(boston_df["medv"], color= "red", bins= 50)

plt.xlabel("Prices of House in Boston")

plt.ylabel("Number of Houses")

plt.title("Price Distribution of House in Boston")

plt.show()
```

```python
# Scatter Plot of House Price vs Number of Rooms:
```

```python
plt.figure(figsize= (10,8))

plt.scatter(boston_df["rm"], boston_df["medv"])

plt.xlabel("Number of Rooms")

plt.ylabel("Price of House")

plt.title("Number of Rooms Vs Prices of House")

plt.show()
```

### 3) Heat Map:

#### Better way to Visualize Two-Dimensional Data

#### Advantages of Heat Map:
#### 1) Draws Attention to Risk Prone Area.
#### 2) Uses entire dataset to draw insights.
#### 3) Used for Cluster Analysis and can be used with Large Datasets.

### Demo: Heat Map

```python
import seaborn as sns
```

```python
flight_df = sns.load_dataset("flights")
```

```python
flight_df.head()
```

```python
# Creating a Pivot table:
```

```python
pivot_tab_df = flight_df.pivot(index= "month", columns= "year", values= "passengers")
```

```python
pivot_tab_df
```

```python
# Heatmap of Pivot Data
```

```python
plt.figure(figsize= (15,12))

sns.heatmap(pivot_tab_df, annot= True)

plt.xlabel("Year", fontsize= 15)

plt.ylabel("Month", fontsize= 15)

plt.title("Year and Month Wise Number of Passengers", fontsize= 20)

plt.show()
```

### -----------------------------------------------------------------------------------------------------

### 4) Pie Chart:

#### Used to Show Percentage or Proportional Data.
#### can be created using pie method.

#### Advantages of Pie Chart:
#### 1) Summarizes large dataset in Visual Form.
#### 2) Displays relative proportions of multiple classes of Data.
#### 3) Size of the Circle is made proportional to total Quantity.

### Demo: Pie Chart

```python
job_data = [40,20,17,8,5,10]
```

```python
labels = ["IT","Finance","Marketing","Admin","HR","Operations"]
```

```python
explode = [0.05,0.02,0.01,0,0,0]
```

```python
plt.figure(figsize= (10,18))

plt.pie(x= job_data, labels= labels, explode= explode, autopct=f"%.2f")

plt.title("No. of Employees is Departments")

plt.show()
```

### -----------------------------------------------------------------------------------------------------

### 5) Error Bar:

#### Used to graphically represent variability od Data.
#### Used mainly to point out Errors.

#### Advantags of Error Bar:
#### 1) Shows Variability in Data and Indicates Errors.
#### 2) Depicts the precision in Data Analysis.
#### 3) Depicts how well a function and model are used in Data Analysis.

### Demo: Error Bar

```python
xval = np.arange(0.1, 4, 0.5)
```

```python
yval = np.exp(-xval)
```

```python
plt.figure(figsize= (15,12))

plt.errorbar(xval, yval, xerr = 0.4, yerr = 0.5)
  
plt.title('Error Bar Example')

plt.show()
```
