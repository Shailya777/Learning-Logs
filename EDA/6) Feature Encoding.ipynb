# Feature Encoding:

Feature encoding is the process of converting categorical data (non-numeric) into a numerical format that machine learning algorithms can understand and process. This transformation is necessary because most algorithms work with numerical inputs.

Common techniques include One-Hot Encoding for nominal data and Ordinal Encoding for data with inherent order, while other methods like Target-Mean Encoding and Hashing Encoding offer different ways to represent categorical features.

#### Why Feature Encoding is Needed
- Model Compatibility: Machine learning models like linear regression and neural networks require numerical input to perform calculations and learn patterns.
- Data Representation: Encoding converts text-based categories (e.g., "Red," "Blue," "Green") into numerical values that the model can interpret meaningfully.

#### Types of Categorical Data and Encoding
- Nominal Data: Data with discrete values and no inherent order (e.g., countries, colors).
1) One-Hot Encoding: Creates new binary columns for each category, with a '1' in the column for the corresponding category and '0' elsewhere.
2) Hashing Encoding: Uses a hashing function to map categories to a fixed number of dimensions, which can be efficient for high-cardinality features. 

- Ordinal Data: Data that has a natural order or ranking (e.g., "Low," "Medium," "High" or education levels).
1) Ordinal Encoding: Assigns a unique integer to each category based on its inherent order, ensuring that the model understands the hierarchy.

#### Advanced Encoding Techniques
- Target-Mean Encoding: Replaces a category with the mean of the target variable for that category, incorporating information about the target into the feature. 
- Frequency Encoding: Replaces categories with their frequency or count in the dataset. 

#### Choosing the Right Technique
The best encoding method depends on the type of categorical data, the number of unique categories, and the specific machine learning algorithm being used.

#### 1) Label Encoding:

- Assigns a unique integer to each category in a feature.

- Suitable for ordinal data where there's an inherent order or ranking among categories (e.g., "low," "medium," "high").

```python
le = LabelEncoder()
df['Gender_Encoded'] = le.fit_transform(df['Gender'])
df.head()
```

```python
df['Gender_Encoded'].value_counts()
```

```python
le = LabelEncoder()
df['Geography_Encoded'] = le.fit_transform(df['Geography'])
df.head()
```

#### 2) One-Hot Encoding:

- Converts each category into a new binary column (0 or 1).
- Suitable for nominal data where there's no inherent order among categories (e.g., "red," "blue," "green").
- Avoids implying any ordinal relationship that doesn't exist.

##### 1) using pd.get_dummies:

```python
df_dummies = pd.get_dummies(data= df, columns= ['Gender', 'Geography'], drop_first= True)
df_dummies.head()
```

##### 2) Using Sklearn:

```python
ohe = OneHotEncoder(sparse_output= False, drop= 'first')
gender_mat = ohe.fit_transform(df[['Gender']])
gender_mat
```

```python
gender_df= pd.DataFrame(data= gender_mat, columns= ohe.get_feature_names_out(['Gender']))
gender_df.head()
```

```python
new_df = df.join(gender_df)
new_df.head()
```

#### Choosing the Right Encoding Method:
- Ordinal data: Label Encoding or Ordinal Encoding (if order needs to be explicit).
- Nominal data: One-Hot Encoding.
- Consider the number of unique categories and potential for high dimensionality with One-Hot Encoding.
- For target-guided encoding (e.g., Mean Encoding, Leave One Out Encoding), specialized libraries like category_encoders can be used.
