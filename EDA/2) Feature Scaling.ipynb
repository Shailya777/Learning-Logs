# Feature Scaling:

Feature scaling is a machine learning preprocessing technique that standardizes numerical features to a common scale, preventing features with larger magnitudes from disproportionately influencing model outcomes and improving algorithm performance. 

Common scaling methods include Min-Max Scaling or Normalization, which rescales data to a specific range like 0 to 1, and Standardization (z-score scaling), which transforms features to have a mean of 0 and a standard deviation of 1. 

This process is vital for algorithms relying on distance calculations, such as k-means clustering, and for faster convergence in gradient-based optimization.

#### Why is Feature Scaling Necessary?
- Prevents Feature Dominance: Without scaling, features with large numerical values (e.g., income in thousands) can dominate those with smaller values (e.g., age in tens), even if the smaller-magnitude features are equally or more important to the model.

- Improves Algorithm Performance: Many machine learning algorithms, particularly those based on distance calculations (like SVM or KNN), are sensitive to feature scales. Scaling ensures that these algorithms can provide more accurate and reliable results. 

- Faster Convergence: For gradient-based algorithms, feature scaling can lead to faster convergence, as the optimization process can avoid the steep, elongated contours of the error surface that result from unscaled features.

### 1) Normalization (Min-Max Scaling):

- How it works: Rescales features to a fixed range, most commonly between 0 and 1. 
- Formula: X_{scaled} = (X - X_min) / (X_max - X_min).
- Best for: Algorithms sensitive to the range of the features, like neural networks, or when a fixed range is desired. 
- Drawback: Can be sensitive to outliers, as the minimum and maximum values are used for the scaling. 

```python
from sklearn.preprocessing import MinMaxScaler
```

##### Small Example:

```python
data = np.array([[2],[3],[5],[6],[6]])
print(data)
print(data.ndim)
print(data.shape)
```

```python
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(data)
print(scaled_data)
```

##### Normalization of Features in Dataframe:

```python
new_df.head()
```

```python
scaler = MinMaxScaler()
scaled_df = scaler.fit_transform(new_df)
print(scaled_df)
print(type(scaled_df))
```

```python
# Turning Returned Array into Dataframe:

scaled_df = pd.DataFrame(data= scaled_df, columns= new_df.columns)
scaled_df.head()
```

```python
print(scaled_df['Age'].max())
print(scaled_df['Age'].min())
print(scaled_df['Age'].mean())
print(scaled_df['Age'].std())
```

```python
print(scaled_df['Tenure'].max())
print(scaled_df['Tenure'].min())
print(scaled_df['Tenure'].mean())
print(scaled_df['Tenure'].std())
```

### 2) Standardization (z-Score Scaling):

- How it works: Transforms features to have a mean of 0 and a standard deviation of 1. 
- Formula: X_{scaled} = (X - \mu)\ sigma, where mu is the mean and sigma is the standard deviation. 
- Best for: Algorithms that assume a normally distributed dataset or require zero-centered features, such as Principal Component Analysis (PCA).
- Benefit: Less affected by outliers compared to Min-Max Scaling. 

```python
from sklearn.preprocessing import StandardScaler
```

##### Small Example:

```python
data = np.array([[2],[3],[5],[6],[6]])
print(data)
print(data.ndim)
print(data.shape)
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
print(scaled_data)
```

##### Standardization of Features in Dataframe:

```python
new_df.head()
```

```python
scaler = StandardScaler()
scaled_df = scaler.fit_transform(new_df)
scaled_df
```

```python
# Turning Returned Array into Dataframe:

scaled_df = pd.DataFrame(data= scaled_df, columns= new_df.columns)
scaled_df.head()
```

```python
print(scaled_df['Age'].max())
print(scaled_df['Age'].min())
print(scaled_df['Age'].mean())
print(scaled_df['Age'].std())
```

```python
print(scaled_df['Tenure'].max())
print(scaled_df['Tenure'].min())
print(scaled_df['Tenure'].mean())
print(scaled_df['Tenure'].std())
```
