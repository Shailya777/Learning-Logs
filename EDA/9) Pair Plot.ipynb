## Assignment 1 - Pair Plot:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("auto_data_mpg.csv")
```

```python
# Changing Origin Values:
# 1 = India
# 2 = USA
# 3 = Australia
```

```python
df["origin"] = df["origin"].map({1 : "India", 2 : "USA", 3 : "Australia"})
```

```python
df["origin"]
```

```python
df["origin"].value_counts()
```

```python
plt.figure(figsize= (15,12))

sns.pairplot(data= df, vars=["mpg","weight"]  ,hue= "origin")

plt.show()
```

```python
plt.figure(figsize= (10,8), dpi= 200)

sns.pairplot(data= df, vars=["mpg","weight"]  ,hue= "origin", kind= "hist")

plt.show()
```

```python
plt.figure(figsize= (10,8))

sns.pairplot(data= df, vars=["mpg","weight"]  ,hue= "origin", kind= "kde")

plt.show()
```

```python
plt.figure(figsize= (10,8))

sns.pairplot(data= df, vars=["mpg","weight"]  ,hue= "origin", kind= "reg")

plt.show()
```
