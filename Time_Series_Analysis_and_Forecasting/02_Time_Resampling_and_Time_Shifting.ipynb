### Pandas Time Series: Resampling & Shifting

- This notebook covers two fundamental techniques for manipulating time series data in Pandas:
    - Resampling: Changing the frequency of your data (e.g., converting daily data to monthly).
    - Shifting: Moving data points backward or forward in time (e.g., comparing today's price to yesterday's).

#### Time Resampling:

- Concept: Think of resample() as a groupby operation for time. It allows you to change the frequency of your time series data.There are two main types of resampling
    - :Downsampling: Converting high frequency to low frequency (e.g., Daily $\rightarrow$ Monthly). You must aggregate the data (mean, sum, max, etc.) because you are squashing multiple days into one month.
    - Upsampling: Converting low frequency to high frequency (e.g., Daily $\rightarrow$ Hourly). This creates gaps in data that usually need to be filled (interpolated).

When calling `.resample()` you first need to pass in a **rule** parameter, then you need to call some sort of aggregation function.

The **rule** parameter describes the frequency with which to apply the aggregation function (daily, monthly, yearly, etc.)<br>
It is passed in using an "offset alias" - refer to the table below. [[reference](http://pandas.pydata.org/pandas-docs/stable/user_guide/timeseries.html#offset-aliases)]

The aggregation function is needed because, due to resampling, we need some sort of mathematical rule to join the rows (mean, sum, count, etc.)

##### 1) Year wise Average:

```python
df.resample(rule= 'YE').mean()
```

```python
df.resample(rule= 'YE').mean()['Close'].plot(kind= 'bar')
```

##### 2) Month wise Total Volume:

```python
df.resample(rule= 'ME').sum()['Volume'].plot(kind= 'bar', figsize= (15,6), title= 'Month wise Total Volume')
```

##### 3) Quarter wise Average Closing Price:

```python
df.resample(rule= 'QE').mean()['Close'].plot(kind= 'bar', figsize= (15,6), title= 'Quarter wise Average Closing Price')
```

#### Time Shifting:

- Concept: Shifting is moving data up or down along the time index. It is crucial for calculating differences over time (e.g., "Day-over-Day" change).

    - A. Shifting Data (Lagging): This moves the value column down, leaving the index (dates) where it is.
      - Positive value (e.g., 1): Shifts data down (creating a "lag"). The first row becomes NaN.
      - Negative value (e.g., -1): Shifts data up (creating a "lead"). The last row becomes NaN.

    - B. Shifting the Time Index:
        - Sometimes you don't want to move the data, you want to move the timestamps. You can do this by passing the freq argument to shift.
        - Use Case: Your data was recorded with the wrong timezone or needs to be pushed forward by a month for forecasting alignment.

```python
df.head()
```

```python
df.tail()
```

```python
df.shift(periods= 1)
```

```python
df.shift(periods= 5)
```

```python
df.shift(periods= -1)
```

`We can choose to shift index values up or down without realigning the data by passing in a freq argument.
This method shifts dates to the next period based on a frequency code. Common codes are 'M' for month-end and 'A' for year-end.`

```python
df
```

```python
df.shift(periods= 1, freq= 'ME')
```

```python
df.shift(periods= 1, freq= 'YE')
```
