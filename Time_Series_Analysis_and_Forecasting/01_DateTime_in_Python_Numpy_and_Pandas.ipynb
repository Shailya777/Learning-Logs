#### 2) Numpy DateTime:

NumPy handles dates more efficiently than Python's datetime format.

The NumPy data type is called datetime64 to distinguish it from Python's datetime.

```python
import numpy as np
```

```python
x = np.array(['2026-01-01', '2026-01-02', '2026-01-03'])
```

```python
print(x)
print(x.dtype)
```

```python
x = np.array(['2026-01-01', '2026-01-02', '2026-01-03'], dtype= 'datetime64')
```

```python
print(x)
print(x.dtype)
```

`[D] shows that the Dates are being stored at the precision of Days.`

```python
# Hour Level Precision:

y = np.array(['2026-01-01', '2026-01-02', '2026-01-03'], dtype= "datetime64[h]")
```

```python
print(y)
print(y.dtype)
```

```python
# Year Level Precision:

z = np.array(['2026-01-01', '2026-01-02', '2026-01-03'], dtype= "datetime64[Y]")
```

```python
print(z)
print(z.dtype)
```

```python
# Month Level Precision:

np.array(['2026-01-01', '2026-01-02', '2026-01-03'], dtype= "datetime64[M]")
```

```python
# Minute Level Precision:

np.array(['2026-01-01', '2026-01-02', '2026-01-03'], dtype= "datetime64[m]")
```

```python
# Second Level Precision:

np.array(['2026-01-01', '2026-01-02', '2026-01-03'], dtype= "datetime64[s]")
```

```python
# Mili Second Level Precision:

np.array(['2026-01-01', '2026-01-02', '2026-01-03'], dtype= "datetime64[ms]")
```

```python
# Nano Second Level Precision:

np.array(['2026-01-01', '2026-01-02', '2026-01-03'], dtype= "datetime64[ns]")
```

`Just as np.arange(start,stop,step) can be used to produce an array of evenly-spaced integers, we can pass a dtype argument to obtain an array of dates.`

```python
np.arange('2026-01-01', '2026-01-31', 3, dtype= 'datetime64[D]')
```

`As we have provided "datetime64[D]" in dtype, 7 steps will be taken in the unit of days.`

```python
np.arange('2026-01-01', '2026-01-07', 7, dtype= 'datetime64[h]')
```

`As we have provided "datetime64[h]" in dtype, 7 steps will be taken in the unit of hours.`

```python
np.arange('2010', '2050', 5, dtype= 'datetime64[Y]')
```

```python
np.arange('2010', '2050', dtype= 'datetime64[Y]')
```

#### 3) Pandas DateTime:

```python
import pandas as pd
```

##### 1) Date range:

```python
pd.date_range(start= '2026-01-01', end= '2026-01-31', freq= 'D')
```

```python
pd.date_range(start= '2026-01-01', end= '2026-01-31', periods= 10)
```

```python
pd.date_range(start= '2026-01-01', periods= 10, freq= 'D')
```

##### 2) to_datetime:

```python
pd.to_datetime(['Jan 01, 2026', '2026-11-10', '01-Feb-1992', '01/11/2026'], format= 'mixed')
```

`Pandas can Identify many formats for  Dates as shown above.
If we want to convert some specific format of Date that Pandas can not recognize, we can specify format exclusively.`

```python
pd.to_datetime(['12-10-1956', '14-06-1998'], format= "%d-%m-%Y")
```

```python
pd.to_datetime(['12--10--1956', '14--06--1998'], format= "%d--%m--%Y")
```

```python
pd.to_datetime(['12--10--90', '14--06--98'], format= "%d--%m--%y")
```
