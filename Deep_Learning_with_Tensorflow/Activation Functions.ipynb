# ⚡ Activation Functions

## 1. The Core Concept
An Activation Function is a mathematical "gate" that determines the output of a neural network node given an input or set of inputs.

### The "Why" - Non-Linearity
This is the most critical interview question: *Why do we need activation functions?*
* **Without them:** $Output = W \cdot X + b$. No matter how many layers you stack, the composition of linear functions is just another linear function. The network could only learn straight lines.
* **With them:** We introduce curves, bends, and complexity. This allows the network to learn complex mappings (e.g., pixel intensities $\rightarrow$ "Cat").

## 2. Major Activation Functions

### A. Step Function
* **Logic:** If input > 0, output 1. Otherwise 0.
* **Pros:** Simple binary classification.
* **Cons:** Not differentiable at 0 (cannot use with Backpropagation). Not used in modern Deep Learning.

### B. Linear Function
* **Equation:** $f(x) = x$
* **Pros:** Good for simple regression outputs.
* **Cons:** No non-linearity. Impossible to use backpropagation effectively in hidden layers.

### C. Sigmoid (Logistic) Function
Historically popular, but rarely used in hidden layers today.
* **Equation:** $$\sigma(z) = \frac{1}{1 + e^{-z}}$$
* **Range:** $(0, 1)$
* **Visual:** S-shaped curve.
* **Use Case:** Output layer for Binary Classification (Probability).
* **Major Flaw (Vanishing Gradient):**
    * The derivative of Sigmoid is $\sigma(z) \cdot (1 - \sigma(z))$.
    * Max value of derivative is 0.25.
    * In a deep network, multiplying many small gradients (0.25 * 0.25 * ...) results in a number close to zero. The network stops learning.





### D. Tanh (Hyperbolic Tangent)
A scaled version of Sigmoid.
* **Equation:** $$\tanh(z) = \frac{e^z - e^{-z}}{e^z + e^{-z}}$$
* **Range:** $(-1, 1)$
* **Pros:** It is **Zero-Centered**. (Unlike Sigmoid, where outputs are always positive, Tanh allows for negative outputs, making optimization easier).
* **Cons:** Still suffers from Vanishing Gradient for very large/small inputs.




### E. ReLU (Rectified Linear Unit)
The default choice for hidden layers in modern Deep Learning.
* **Equation:** $$R(z) = \max(0, z)$$
* **Logic:** If input is positive, output the input. If negative, output 0.
* **Derivative:** 1 if $z > 0$; 0 if $z < 0$.
* **Pros:**
    * **Computationally Cheap:** No exponents, just a `max()` check.
    * **Solves Vanishing Gradient:** For positive inputs, the gradient is always 1. It doesn't shrink.
* **Cons (The "Dying ReLU"):**
    * If a neuron falls into the negative side, its output is 0 and its gradient is 0. It becomes inactive and never recovers.




### F. Leaky ReLU
A patch for the "Dying ReLU" problem.
* **Equation:** $$f(z) = \max(\alpha z, z)$$ where $\alpha$ is small (e.g., 0.01).
* **Logic:** Instead of flatlining at 0 for negative inputs, it has a slight slope. This ensures gradients still flow (the neuron keeps learning) even for negative values.



### G. Softmax
Exclusively used in the **Output Layer** for Multi-Class Classification.
* **Function:** Generalizes the Sigmoid function to $K$ classes.
* **Logic:** Converts a vector of raw numbers (logits) into a probability distribution.
* **Key Property:** The sum of all outputs equals 1.
    $$P(y=j) = \frac{e^{z_j}}{\sum_{k=1}^{K} e^{z_k}}$$
