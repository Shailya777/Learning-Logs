# 🧠 Introduction to Deep Learning

## 1. What is Deep Learning?
Deep Learning is a specialized subset of Machine Learning inspired by the structure and function of the human brain. While traditional Machine Learning algorithms (like Linear Regression, SVM, or Random Forest) often rely on manual feature engineering, Deep Learning automates this process.

It uses multi-layered structures called **Artificial Neural Networks (ANNs)** to extract high-level features from raw data (images, sound, text) in a hierarchical manner.

### The Hierarchy of Learning
* **Artificial Intelligence:** The broad discipline of creating intelligent machines.
* **Machine Learning:** A subset of AI where machines learn from data without explicit programming.
* **Deep Learning:** A subset of ML based on deep neural networks.

### Why "Deep"?
The word "Deep" refers to the number of **hidden layers** in the network.
* **Shallow Network:** Input + 1 Hidden Layer + Output.
* **Deep Network:** Input + Multiple Hidden Layers + Output.
* **Universal Approximation Theorem:** A theoretical proof stating that a neural network with just one hidden layer can approximate *any* continuous function, given enough neurons. Deep Learning takes this further by stacking layers to learn complex, non-linear patterns efficiently.

## 2. History & Evolution
* **1943 (McCulloch-Pitts Neuron):** The first mathematical model of a biological neuron. It was a simple logic gate (Binary Input $\rightarrow$ Binary Output).
* **1957 (The Perceptron):** Frank Rosenblatt invented the Perceptron, a hardware algorithm for image recognition. It could learn weights but failed at non-linear problems (the XOR problem).
* **1986 (Backpropagation):** Geoffrey Hinton, David Rumelhart, and Ronald Williams popularized the Backpropagation algorithm, allowing multi-layer networks to actually learn.
* **2006 (Deep Belief Networks):** Hinton introduced "greedy layer-wise pre-training," proving deep networks could be trained effectively.
* **2012 (AlexNet - The Boom):** The pivotal moment. Alex Krizhevsky used a Deep Convolutional Neural Network (CNN) and GPUs to crush the ImageNet competition, reducing error rates from 26% to 15%. This started the modern AI era.

## 3. The Biological Inspiration vs. Artificial Reality
We model ANNs after the brain, but they are mathematical simplifications, not exact replicas.

| Biological Neuron | Artificial Neuron (Perceptron) | Role |
| :--- | :--- | :--- |
| **Dendrites** | **Input Layer ($X$)** | Receives signals/data. |
| **Synapse** | **Weights ($W$)** | Controls the strength of the connection. |
| **Soma (Cell Body)** | **Summation ($\sum$)** | Aggregates inputs. |
| **Threshold** | **Bias ($b$)** | Shifts the activation boundary. |
| **Firing Rate** | **Activation Function** | Decides if the neuron is active. |
| **Axon** | **Output ($y$)** | Transmits the result to the next layer. |

## 4. The Mathematical Model of a Neuron
A single neuron performs a 2-step calculation:

1.  **Weighted Sum (Linear):**
    $$Z = (x_1 \cdot w_1) + (x_2 \cdot w_2) + ... + (x_n \cdot w_n) + b$$
    * $x$: Inputs
    * $w$: Weights (The "importance" of the input)
    * $b$: Bias (Allows the neuron to shift the activation function left or right)

2.  **Activation (Non-Linear):**
    $$A = f(Z)$$
    * If we don't apply $f(Z)$, the entire network remains just a Linear Regression model.
