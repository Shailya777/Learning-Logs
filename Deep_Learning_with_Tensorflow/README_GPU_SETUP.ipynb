## Guide: Setting up TensorFlow with GPU (RTX 4080) on Windows via WSL2:

Author: Shailya

Date: January 2026

System: Windows 11 | Intel i9-13900HX | NVIDIA RTX 4080 (12GB)

This document details the step-by-step process to bypass native Windows GPU limitations by setting up WSL2 (Windows Subsystem for Linux), installing the NVIDIA drivers, and configuring a robust Deep Learning environment.

#### Phase 1: Installing the Linux Subsystem
Goal: Enable the Linux kernel inside Windows to run high-performance AI tools.

1. Install WSL2 (Run in Windows PowerShell as Admin):

   `wsl --install`

What this does: Downloads the necessary Linux kernel components and installs the default distribution (Ubuntu). A system restart is required after this step.

2. Initialize Ubuntu:
After restart, open the "Ubuntu" app from the Start menu.
Create a username and password when prompted.

`----------------------------------------------------------------------------------------------------------------------`

#### Phase 2: Setting up the Python Manager (Miniconda)
Goal: Install conda to manage Python versions and environments cleanly, separate from the system Python.

1. Download the Installer:

   `mkdir -p ~/miniconda3`

    `wget https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh -O ~/miniconda3/miniconda.sh`

What this does: Creates a directory and fetches the latest Linux version of Miniconda from the web.

2. Run the Installer:

   `bash ~/miniconda3/miniconda.sh -b -u -p ~/miniconda3`

What this does: Executes the installation script in "batch" mode (-b), so it doesn't ask for manual confirmations.

3. Cleanup and Initialization:

   `rm -rf ~/miniconda3/miniconda.sh`

   `~/miniconda3/bin/conda init bash`

What this does: Deletes the installer file to save space and adds Conda to the system path (.bashrc) so the conda command works every time we open the terminal. (Restart terminal after this step).

`----------------------------------------------------------------------------------------------------------------------`

#### Phase 3: Creating the Deep Learning Environment
Goal: Create an isolated environment with Python 3.11 and the necessary GPU drivers.

1. Accept Anaconda Terms of Service (Necessary to download packages without interruption):

   `conda tos accept --override-channels --channel https://repo.anaconda.com/pkgs/main`

   `conda tos accept --override-channels --channel https://repo.anaconda.com/pkgs/r`

2. Create the Environment:

   `conda create --name tf-gpu python=3.11 -y`

What this does: Creates a sandbox named tf-gpu running Python 3.11 (a stable version for ML libraries).

3. Activate the Environment:

   `conda activate tf-gpu`

What this does: Switches the shell to use our new environment. You should see (tf-gpu) on the command line.

`----------------------------------------------------------------------------------------------------------------------`

#### Phase 4: Installing TensorFlow & Data Science Stack
Goal: Install the AI framework (TensorFlow) and standard data analysis tools.

1. The "Magic" TensorFlow Install:
   
   `pip install "tensorflow[and-cuda]"`

What this does: This is the critical step. The [and-cuda] tag tells pip to download not just TensorFlow, but also the compatible NVIDIA CUDA and cuDNN libraries automatically. This ensures the GPU drivers match the software version perfectly.

2. Install Data Science Libraries:

   `pip install pandas scikit-learn matplotlib seaborn jupyterlab jupyterlab-git`

What this does: Installs
- pandas/numpy: Data manipulation.
- matplotlib/seaborn: Plotting and graphs.
- scikit-learn: Standard Machine Learning algorithms.
- jupyterlab: The notebook interface.
- jupyterlab-git: GUI for Git integration inside Jupyter.

`----------------------------------------------------------------------------------------------------------------------`

#### Phase 5: Verification
Goal: Confirm the GPU is actually detected.

Run this Python script inside the terminal:

`import tensorflow as tf`

`print("Num GPUs Available: ", len(tf.config.list_physical_devices('GPU')))`

Expected Output: Num GPUs Available: 1

`----------------------------------------------------------------------------------------------------------------------`

#### Phase 6: Git & Project Setup
Goal: Clone the existing repository and enable pushing to GitHub.

1. Configure Git (Link to Windows Credentials):
   
   `git config --global credential.helper "/mnt/c/Program\ Files/Git/mingw64/libexec/git-core/git-credential-manager.exe"`

What this does: Tells the Linux Git to use the Windows Credential Manager. This means you don't need to generate new SSH keys; it "borrows" your Windows login.

2. Configure User Details:

   `git config --global user.name "Your Name"`
   
   `git config --global user.email "your.email@example.com"`

3. Clone Repository:

   `git clone https://github.com/YOUR_USERNAME/YOUR_REPO.git`
