# Hyperparameter Optimization:

## 1. What is Hyperparameter Optimization (HPO)?

In Machine Learning, a model has two types of parameters:
* **Model Parameters:** These are learned from the data during training (e.g., the weights in linear regression or split points in a decision tree).
* **Hyperparameters:** These are set *before* the learning process begins. They control the behavior of the training algorithm (e.g., the learning rate, the depth of a tree, or the number of neighbors in KNN).

**Hyperparameter Optimization** is the process of finding the optimal set of hyperparameters that results in the best model performance (usually minimizing loss or maximizing accuracy).

## 2. Why is it Necessary?

Default hyperparameters in libraries like Scikit-Learn are chosen to work "okay" on average, but they are rarely optimal for your specific dataset. Proper HPO can:
* Prevent overfitting or underfitting (bias-variance trade-off).
* Significantly improve accuracy/F1-score.
* Reduce training time by finding efficient convergence parameters.

## 3. Common Methods of HPO

### A. Manual Search
The practitioner sets hyperparameters based on intuition or past experience.
* *Pros:* Good for gaining intuition.
* *Cons:* Highly inefficient, hard to reproduce, and unlikely to find the global optimum.

### B. Grid Search
An exhaustive search over a manually specified subset of the hyperparameter space.
* *Pros:* Guaranteed to find the best combination within the specified grid.
* *Cons:* Computationally expensive. The cost grows exponentially with the number of hyperparameters (Curse of Dimensionality).

### C. Randomized Search
Parameters are sampled from a defined distribution (e.g., uniform, log-uniform) for a fixed number of iterations.
* *Pros:* More efficient than Grid Search. It is statistically likely to find a better model in less time because it explores the continuous space more effectively.
* *Cons:* Not guaranteed to find the absolute best combination (though usually finds a "good enough" one).

### D. Bayesian Optimization
A probabilistic model-based approach. It builds a probability model (surrogate model) of the objective function and uses it to select the most promising hyperparameters to evaluate next.
* *Pros:* Very efficient. It "learns" from previous iterations, focusing on promising areas of the hyperparameter space.
* *Cons:* Harder to parallelize than Grid/Random search.

### E. Other Methods
* **Halving Grid/Random Search:** Iteratively trains on small subsets of data and kills "loser" configurations early (Successive Halving).
* **Evolutionary Algorithms:** Uses genetic algorithms to mutate and cross-over hyperparameter sets.
