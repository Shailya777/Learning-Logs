# Feature Encoding:

Feature encoding is a core process in machine learning that converts categorical data—non-numeric values like text or labels—into a numerical format that algorithms can understand and process. The specific encoding technique used depends on the nature of the data and the requirements of the machine learning model.

- Common encoding techniques:
    -  One-hot encoding: This method represents each category as a binary vector.
       - How it works: For a feature with N unique categories, it creates N new binary columns. For a given data point, the column corresponding to its category gets a value of 1, while all other new columns get a value of 0.
       - Best for: Nominal (unordered) categorical data with low to medium cardinality (few unique categories).Example: A "Color" feature with categories "Red," "Green," and "Blue" would be converted into three columns: "Color_Red," "Color_Green," and "Color_Blue.
       - "Considerations: It can lead to a "curse of dimensionality" and increased training time if a feature has many unique categories (high cardinality).
    
    - Label encoding: This simple method assigns a unique integer to each category.
         - How it works: Categories are mapped to integers, such as "Red" = 0, "Green" = 1, and "Blue" = 2.
         - Best for: Ordinal (ordered) categorical data, like "small," "medium," and "large".
         - Considerations: Applying label encoding to nominal data can mislead a model into assuming an ordinal relationship exists where it doesn't, which is especially problematic for linear models.
    
    - Ordinal encoding: Similar to label encoding, this method maps each category to an integer, but the mapping is specified manually to preserve the inherent order of the data.
         - How it works: The user provides the correct order for the categories to ensure the numerical representation aligns with the data's ranking, for example, "High School" = 0, "Bachelor's" = 1, "Master's" = 2.
         - Best for: Ordinal data where a clear ranking exists.
         - Considerations: Incorrectly specifying the order can negatively impact model performance.
    
    - Target encoding (Mean encoding): This supervised technique replaces each category with the mean of the target variable for that category.
         -  How it works: For a classification problem, a category is replaced by the average target value (e.g., probability of being in the positive class) for all instances of that category.
         - Best for: High-cardinality nominal features, as it doesn't increase the feature space.
         - Considerations: This method is prone to data leakage and overfitting, so it requires validation strategies like using cross-validation or adding smoothing to the mean calculation.
    
    - Frequency (or count) encoding: This method replaces each category with the frequency or count of its occurrence in the dataset.
         -  How it works: A category is mapped to the number of times it appears in the dataset. For instance, if "USA" appears 500 times in a "Country" column, it is encoded as 500.
         - Best for: Datasets where the frequency of a category is predictive of the target variable.
         - Considerations: Can cause collisions if different categories have the same frequency.
    
    - Binary encoding: This technique encodes categories by converting them to a binary representation.
         - How it works: It first assigns an integer to each category (like label encoding) and then converts those integers into binary code. The digits of the binary code are then split into separate columns.
         - Best for: High-cardinality variables, as it reduces dimensionality compared to one-hot encoding.
         - Considerations: The encoded features are less interpretable than one-hot encoded features.

- How to choose the right encoding method: The choice of encoding depends on several factors:

    - Type of variable: Is the data nominal (no order) or ordinal (ordered)?
    - Cardinality: How many unique categories does the feature have? High-cardinality features can cause issues with one-hot encoding.
    - Algorithm: Some algorithms, like linear models, can be sensitive to false ordinal relationships, making one-hot encoding a safer choice for nominal data. Tree-based models can often handle label or ordinal encoded data without issue.
    - Interpretability: Some encoding methods, like target encoding, are harder to interpret than simple one-hot encoding.
    - Risk of overfitting: Techniques that use the target variable for encoding, such as target encoding, can increase the risk of overfitting if not handled properly.

#### Example:

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, LabelEncoder, OrdinalEncoder
```

##### 1) One Hot Encoding:

###### 1) Using Scikit Learn Library:

```python
df = pd.DataFrame({'Color': ['Red', 'Blue', 'Green', 'Blue', 'Green', 'Green',
                            'Red', 'Green', 'Blue', 'Blue']})
```

```python
df
```

```python
df.value_counts()
```

```python
encoder = OneHotEncoder(sparse_output= False)
```

```python
encoded_df = encoder.fit_transform(df[['Color']])
```

```python
encoded_df
```

```python
encoder.get_feature_names_out(['Color'])
```

```python
encoded_df = pd.DataFrame(encoded_df, index= df.index, columns= encoder.get_feature_names_out(['Color']))
encoded_df
```

###### 2) Using Pandas Library:

```python
df = pd.DataFrame({'Color': ['Red', 'Blue', 'Green', 'Blue', 'Green', 'Green',
                            'Red', 'Green', 'Blue', 'Blue']})
```

```python
df
```

```python
df.value_counts()
```

```python
df_dummies = pd.get_dummies(df, columns= ['Color'], dtype= 'int')
```

```python
df_dummies
```

##### 2) Label Encoding:

```python
df = pd.DataFrame({'Size': ['Small', 'Medium', 'Large', 'Large', 'Small',
                           'Medium', 'Medium', 'Large', 'Small', 'Small']})
```

```python
df
```

```python
df.value_counts()
```

```python
l_encoder = LabelEncoder()
```

```python
encoded_df = l_encoder.fit_transform(df['Size'])
```

```python
encoded_df
```

```python
encoded_df = pd.DataFrame(encoded_df, columns= df.columns)
encoded_df
```

##### 3) Ordinal Encoding:

```python
df = pd.DataFrame({'Education': ["High School", "Bachelor's", "Master's",
                                "Bachelor's", "Bachelor's", "Bachelor's",
                                "Master's", "Master's", "High School", "High School"]})
```

```python
df
```

```python
df.value_counts()
```

```python
# Defining Order for The Categories:
category_order = ["High School", "Bachelor's", "Master's"]
```

```python
o_encoder = OrdinalEncoder(categories= [category_order])
```

```python
encoded_df = o_encoder.fit_transform(df[['Education']])
encoded_df
```

```python
encoded_df = pd.DataFrame(encoded_df, columns= df.columns)
encoded_df
```
