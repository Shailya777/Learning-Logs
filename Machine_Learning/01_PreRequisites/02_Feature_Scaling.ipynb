# Feature Scaling:

Feature scaling is a data preprocessing technique that standardizes the range of independent variables (features) in a dataset. This process ensures that all features contribute equally to the model's computations by bringing them to a similar scale, preventing features with larger numerical ranges from dominating others and improving model performance, especially for algorithms sensitive to feature magnitudes, such as gradient descent or distance-based algorithms. Common scaling methods include normalization (scaling to a fixed range, often 0-1) and standardization (scaling to a mean of 0 and a standard deviation of 1). 

- Why is Feature Scaling Important?
    - Prevents Feature Dominance: Without scaling, features with larger values (e.g., salary in thousands) can disproportionately influence the model compared to features with smaller values (e.g., age in tens), leading to poor performance. 
    - Improves Model Convergence: For algorithms like logistic regression and neural networks that use gradient descent, feature scaling helps them converge faster and more reliably. 
    - Enhances Algorithm Performance: Many algorithms, particularly those that rely on distance calculations (like K-Nearest Neighbors or K-Means), perform better when features are on a similar scale.

- Common Feature Scaling Methods:
    - Normalization (Min-Max Scaling): Transforms features to a specific range, usually between 0 and 1.
    - Standardization (Z-score Scaling): Rescales features so that they have a mean of 0 and a standard deviation of 1. This method is less affected by outliers than Min-Max scaling.

- When is Feature Scaling Needed?
    - Gradient-based algorithms: Linear Regression, Logistic Regression, and Neural Networks.
    - Distance-based algorithms: K-Nearest Neighbors (KNN) and Support Vector Machines (SVMs). 
    - Algorithms using Euclidean distance: K-Means Clustering.

- When is it NOT always necessary?
    - Tree-based algorithms: Decision Trees, Random Forests, and XGBoost are generally less affected by feature scaling. 
    - Algorithms that are not sensitive to scale: Some distance-agnostic algorithms.

#### Example:

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
df = pd.read_csv('boston.csv')
```

```python
df.head()
```

```python
df = df.drop('Unnamed: 0', axis= 1)
```

```python
df.head()
```

```python
df.describe().T
```

##### 1) Standardization:

```python
std_scaler = StandardScaler()
df_std_scaled = std_scaler.fit_transform(df)
df_std_scaled
```

```python
df_std_scaled = pd.DataFrame(df_std_scaled, index= df.index, columns= df.columns)
df_std_scaled.head()
```

```python
df_std_scaled.describe().T
```

##### 2) Normalization:

```python
minmax_scaler = MinMaxScaler()
df_minmax_scaled = minmax_scaler.fit_transform(df)
df_minmax_scaled
```

```python
df_minmax_scaled = pd.DataFrame(df_minmax_scaled, index= df.index, columns= df.columns)
df_minmax_scaled.head()
```

```python
df_minmax_scaled.describe().T
```
