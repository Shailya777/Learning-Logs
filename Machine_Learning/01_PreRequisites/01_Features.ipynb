# Features:

#### In machine learning, features and the target variable (also called the label or dependent variable) are fundamental components of supervised learning tasks.

#### Features (X):

- Definition: Features are the individual, independent variables or attributes in a dataset that are used as input to a machine learning model. They represent the characteristics or properties of the data points.

- Role: The model learns patterns and relationships within these features to make predictions or classifications about the target variable.

- Example: In a dataset predicting house prices, features could include the number of bedrooms, square footage, location, and age of the house.

- Representation: Features are typically represented as a matrix or a set of input variables, often denoted as 'X'.

- Feature Engineering: The process of creating new features from existing ones or transforming raw data into a more suitable format for the model is called feature engineering. This can significantly impact model performance.


#### Target Variable (Y):

- Definition: The target variable is the dependent variable that the machine learning model aims to predict or classify. It represents the outcome or output we are interested in.

- Role: In supervised learning, the model learns to map the input features to the corresponding target values during training.

- Example: In the house price prediction example, the target variable would be the actual house price. In a classification problem, it could be a category like "spam" or "not spam."

- Representation: The target variable is typically represented as a vector or a single output variable, often denoted as 'y'.

- Importance: A well-defined and understood target variable is crucial for building effective supervised learning models, as the model's entire purpose revolves around accurately predicting or classifying this variable.

For this house price dataset:
- We have Features / Independent Variables / Attributes / X Variables:
  - CRIM: Per capita crime rate by town
  - ZN: Proportion of residential land zoned for lots over 25,000 sq. ft.
  - INDUS: Proportion of non-retail business acres per town
  - CHAS: Charles River dummy variable (= 1 if tract bounds river; 0 otherwise)
  - NOX: Nitric oxides concentration (parts per 10 million)
  - RM: Average number of rooms per dwelling
  - AGE: Proportion of owner-occupied units built prior to 1940
  - DIS: Weighted distances to five Boston employment centers
  - RAD: Index of accessibility to radial highways
  - TAX: Full-value property-tax rate per $10,000
  - PTRATIO: Pupil-teacher ratio by town
  - B: 1000(Bk − 0.63)^2, where Bk is the proportion of Black residents by town
  - LSTAT: Percentage of lower status of the population.

- We have Target / Dependent Variable / y Variable:
  - Price: Median value of owner-occupied homes in $1000s.
