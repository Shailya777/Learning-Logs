#### 2) Building a vocabulary (Creating a "Bag of Words"):

Let's create dictionaries that correspond to unique mappings of the words in the documents. We can begin to think of this as mapping out all the possible words available for all (both) documents.

```python
# Words from file 1:

with open('One.txt') as one:
    words1 = one.read().lower().split()

with open('Two.txt') as two:
    words2 = two.read().lower().split()
```

```python
print(words1)
print(words2)
```

```python
print(len(words1))
print(len(words2))
```

```python
# Unique words from file 1:
unique_words1 = set(words1)

# Unique words from file 2:
unique_words2 = set(words2)
```

```python
print(unique_words1)
print(unique_words2)
```

```python
# Getting all the words from both files in One Set:

all_unique_words = set()
all_unique_words.update(unique_words1)
all_unique_words.update(unique_words2)

print(all_unique_words)
print(len(all_unique_words))
```

```python
# Creating a Disctionary of all Unique Words:

dict_of_words = dict()
counter = 0

for word in all_unique_words:
    dict_of_words[word] = counter
    counter += 1

print(dict_of_words)
```

#### 3) Bag of Words:

1) We will create a Matrix or DataFrame where columns are all the unique words from all the Documnets.

2) Rows will represent a single document.

3) Cells of Matrx or DataFrame will show how many times a Particular Word (Column) is Present in a Particular Document (Row).

```python
# Creating Emply Lists for Both documents of Same Length as Dictionary of Unique Words:

word_freq1 = [0] * len(dict_of_words)
word_freq2 = [0] * len(dict_of_words)
```

```python
print(word_freq1)
print(word_freq2)
```

```python
# Adding Frequency of Each Word Per Documnet in List:

for word in words1:
    index = dict_of_words[word]
    word_freq1[index] = word_freq1[index] + 1

for word in words2:
    index = dict_of_words[word]
    word_freq2[index] = word_freq2[index] + 1
```

```python
print(word_freq1)
print(word_freq2)
```

```python
# Creating Dataframe:

import pandas as pd

bag_of_words = pd.DataFrame(data= [word_freq1, word_freq2], columns= dict_of_words.keys())

bag_of_words
```
