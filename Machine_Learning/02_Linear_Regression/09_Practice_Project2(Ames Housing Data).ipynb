- Lot Frontage mostly depends on which neighbourhood the house belongs to.
- We can use Neighbourhood feature to find Median Lot Frontage Value to fill in missing values.

```python
class GroupImputer(BaseEstimator, TransformerMixin):
    """
    Group-based imputer for filling missing values in numeric or categorical columns.
    Supports multiple (group_col, target_col) pairs and multiple strategies (mean, median, mode).
    
    Parameters
    ----------
    group_target_pairs : list of tuples
        Each tuple is (group_col, target_col)
        Example: [('Neighborhood', 'LotFrontage'), ('GarageType', 'GarageYrBlt')]

    strategy : str, default='median'
        Imputation strategy. Options: 'median', 'mean', 'mode'
    """
    
    
    def __init__(self, group_target_pairs, strategy='median'):
        self.group_target_pairs = group_target_pairs
        self.strategy = strategy
        self.group_stats_ = {}
        self.global_stats_ = {}

    def fit(self, X, y= None):
        X = X.copy()
        
        for group_col, target_col in self.group_target_pairs:
            if self.strategy == 'median':
                group_values = X.groupby(group_col)[target_col].median()
                global_value = X[target_col].median()
            elif self.strategy == 'mean':
                group_values = X.groupby(group_col)[target_col].mean()
                global_value = X[target_col].mean()
            elif self.strategy == 'mode':
                group_values = X.groupby(group_col)[target_col].agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
                global_value = X[target_col].mode().iloc[0] if not X[target_col].mode().empty else np.nan
            else:
                raise ValueError("strategy must be 'median', 'mean', or 'mode'")

            self.group_stats_[(group_col, target_col)] = group_values
            self.global_stats_[(group_col, target_col)] = global_value
        
        return self

    def transform(self, X):
        X = X.copy()
        
        for group_col, target_col in self.group_target_pairs:
            group_values = self.group_stats_[(group_col, target_col)]
            global_value = self.global_stats_[(group_col, target_col)]

            # Vectorized group-based filling
            X[target_col] = X[target_col].fillna(
                X[group_col].map(group_values)
            ).fillna(global_value)

        return X
```

- This class will help in future to fill in any missing value for a feature based on group mean / median / mode based on another feature.
