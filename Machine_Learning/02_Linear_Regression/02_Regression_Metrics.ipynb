# Regression Metrics:

Regression metrics are statistical measures used to evaluate how well a machine learning model predicts a continuous value (a "target variable") by quantifying the difference between the model's predicted outputs and the actual, observed values.

Common regression metrics include Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R-squared (R²). These metrics help assess model accuracy, compare different models, and identify areas for improvement in a model's predictive performance. 

- Key Regression Metrics:
    - Mean Absolute Error (MAE): The average of the absolute differences between the predicted and actual values. It's useful because it's not heavily influenced by outliers and provides a straightforward measure of error magnitude. 

    - Mean Squared Error (MSE): The average of the squared differences between predicted and actual values. Squaring the errors penalizes larger mistakes more heavily than smaller ones. 

    - Root Mean Squared Error (RMSE): The square root of MSE. This metric is often preferred because it brings the error measure back to the same units as the target variable, making it easier to interpret. 

    - R-squared (R²): Also known as the coefficient of determination, R² measures the proportion of the variance in the dependent variable that is predictable from the independent variables in the model. An R² of 1 indicates the model can explain all the variation, while a lower value suggests a less effective model. 

- Importance:
    - Performance Assessment: Regression metrics provide a quantitative score that shows how accurate the model is. 
    - Model Comparison: They allow you to compare the performance of different regression models on the same dataset to determine which one is superior. 
    - Model Improvement: By analyzing the metrics, data scientists can identify if a model is performing poorly and needs refinement or if additional features are required. 
    - Model Selection: Metrics help in choosing the best model for a specific task by providing a basis for selecting the most suitable one.

#### 1) Mean Absolute Error (MAE):

MAE is the average of the absolute differences between the actual and predicted values. Because it uses absolute values, all individual differences are weighted equally.

- Interpretation: The average error of the model's predictions in the same units as the target variable. A lower value indicates a better-fitting model.

- Best for: When outliers should not disproportionately influence the error. It is more robust to them than MSE.

#### 2) Mean Squared Error (MSE):

MSE is the average of the squared differences between the actual and predicted values. Squaring the differences amplifies the impact of larger errors, which means outliers will have a more significant effect on the final score.

- Interpretation: The average squared prediction error. The unit of MSE is the square of the unit of the target variable, making it less interpretable than MAE or RMSE.

- Best for: As a cost function for optimization algorithms like gradient descent because the error function is convex and differentiable.

#### 3) Root Mean Squared Error (RMSE):

RMSE is the square root of the MSE. It converts the squared error back into the same units as the original target variable, making it more interpretable.

- Interpretation: Represents the average magnitude of the errors, in the same units as the target variable. Like MSE, it is also sensitive to outliers because of the squaring of the residuals.

- Best for: When large errors are particularly undesirable and should be penalized more heavily. It is a good metric to use for reporting model performance to stakeholders.

#### 4) Mean Absolute Percentage Error (MAPE):

MAPE measures the average percentage error of the predictions. It is useful for understanding error in relative terms rather than absolute values.

- Interpretation: The average percentage deviation of the forecast from the actual value.

- Best for: Time-series forecasting and when comparing models on datasets with different scales. It fails, however, if the actual value yt is zero.

#### 5) R-Squared:

R-Squared, or the coefficient of determination, is a statistical measure that represents the proportion of the variance in the dependent variable that is predictable from the independent variables. It provides a measure of how well the model fits the observed data.

- Interpretation: Ranges from 0 to 1. A value of 1 indicates a perfect fit, while 0 indicates the model explains none of the variance. A key limitation is that \(R^{2}\) never decreases when more predictors are added, which can promote overfitting.

- Best for: Simple regression problems to assess overall goodness-of-fit.

#### 6) Adjusted R-Squared:

Adjusted R-Squared is a modified version of R-Squared that accounts for the number of predictors in the model. It increases only if a new predictor improves the model more than would be expected by chance and penalizes the addition of irrelevant predictors.

- Interpretation: A more reliable indicator of a model's predictive power in multiple regression. It can be lower than R-Squared and can even be negative.

-  Best for: Comparing models with different numbers of predictors to avoid overfitting.

#### 7) Mean Squared Logarithmic Error (MSLE):

MSLE measures the ratio between the actual and predicted values by taking the logarithm of each plus one. It is useful when the target variable spans a large range and when you want to penalize under-prediction more than over-prediction.

- Interpretation: A smaller value means a better model fit. Because it measures relative error, it is useful for problems where the error magnitude is proportional to the value of the target.

- Best for: Problems with large variations in the target variable, such as price prediction, where a small absolute error on a high-value item is more important than on a low-value one.
